## **5. Build the Network Communication Graph**

Builds an asset-level graph from the flow records: each unique IP address becomes a **node** (an ICS asset — PLC, sensor, HMI, or SCADA server), and each `sIPs → rIPs` pair that exchanged at least one flow becomes an **edge**. This collapses 45,000+ individual flow rows down to a small topology graph of the ICS testbed itself, which is the structure the GNN will eventually operate on.

Built from `df_graph` (the missing-value-checked version) rather than the raw `df`, so any rows with missing addresses are excluded from the graph.

````python
import networkx as nx

G = nx.from_pandas_edgelist(
    df_graph,
    source='sIPs',
    target='rIPs'
)

print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())
print("\nNodes:")
print(list(G.nodes()))
````

````python
degrees = dict(G.degree())

print("Node degree (number of distinct devices each node talks to):\n")
for node, degree in sorted(degrees.items(), key=lambda x: x[1], reverse=True):
    print(node, degree)
````

**Output:** around 9 nodes and 13 edges — consistent with a small ICSSIM testbed (a handful of PLCs, sensors, an HMI, and a SCADA server). High-degree nodes are the most central devices in the network (typically the SCADA server or a switch/gateway address) and are good early candidates for high-priority asset nodes in the eventual knowledge graph.

## **8. Weighted Graph — Flow Count per Device Pair**

The graph in Section 5 only records *whether* two devices communicated, not *how much*. This section builds a directed, weighted version of the graph, where each edge weight is the number of flows recorded between that source and destination — a more informative input for the eventual GNN, which will use edge weight as a feature.

````python
edge_weights = (
    df_graph.groupby(["sIPs", "rIPs"])
    .size()
    .reset_index(name="flow_count")
)

G_weighted = nx.DiGraph()
for _, row in edge_weights.iterrows():
    G_weighted.add_edge(row["sIPs"], row["rIPs"], weight=row["flow_count"])

print("Weighted graph — Nodes:", G_weighted.number_of_nodes())
print("Weighted graph — Edges:", G_weighted.number_of_edges())
edge_weights.sort_values("flow_count", ascending=False)
````

## **9. Attack-Aware Graph — Highlighting Malicious Edges**

Identifies which device pairs ever carried attack traffic (using `NST_M_Label`) and re-draws the graph with those edges highlighted in red. This is the step that turns a plain network diagram into a security-relevant graph — the same structure the knowledge graph will eventually encode as labelled, risk-weighted edges.

````python
attack_edges = (
    df_graph[df_graph["NST_M_Label"] != "Normal"]
    .groupby(["sIPs", "rIPs", "NST_M_Label"])
    .size()
    .reset_index(name="count")
)
print("Device pairs that carried attack traffic:\n")
print(attack_edges)

# G is undirected, so check both (u, v) and (v, u) — NetworkX does not
# guarantee edges are returned in the same direction they were inserted.
attack_pairs = set(zip(attack_edges["sIPs"], attack_edges["rIPs"]))
edge_colors = [
    "red" if (u, v) in attack_pairs or (v, u) in attack_pairs else "gray"
    for u, v in G.edges()
]

plt.figure(figsize=(8, 6))
nx.draw(
    G,
    pos,
    with_labels=True,
    node_color="lightblue",
    edge_color=edge_colors,
    node_size=1500,
    font_size=8
)
plt.title("ICS-Flow Graph — Red Edges Carried Attack Traffic")
plt.savefig("week2_attack_graph.png", dpi=150, bbox_inches="tight")
plt.show()
````

## **Attack Flow Graph**

````python
# Attack Flow Graph

import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.lines import Line2D
import math

# 1. PREPARE ATTACK FLOWS
attack_timeline = (
    df_graph[df_graph['NST_M_Label'] != 'Normal']
    .groupby(['sIPs', 'rIPs', 'NST_M_Label'])
    .agg(
        count=('NST_M_Label', 'size'),
        first_seen=('startDate', 'min'),
        last_seen=('endDate', 'max')
    )
    .reset_index()
    .sort_values('first_seen')
)

# 2. BUILD MULTIGRAPH
G_path = nx.MultiDiGraph()
for _, row in attack_timeline.iterrows():
    G_path.add_edge(
        row['sIPs'], row['rIPs'],
        attack_type=row['NST_M_Label'],
        count=int(row['count']),
        first_seen=str(row['first_seen'])
    )

# 3. ROLE & COLOR MAPS
role_map = {
    '192.168.0.1':  'Gateway',
    '192.168.0.11': 'PLC1',
    '192.168.0.12': 'PLC2',
    '192.168.0.21': 'HMI1',
    '192.168.0.22': 'HMI2',
    '192.168.0.23': 'HMI3',
    '192.168.0.41': 'Attacker',
    '192.168.0.42': 'C2 Server',
}
attack_colors = {
    'mitm':      '#9333EA',
    'ddos':      '#E24B4A',
    'replay':    '#EF9F27',
    'port-scan': '#378ADD',
}
node_colors_map = {
    'PLC1':      '#1D9E75',
    'PLC2':      '#1D9E75',
    'HMI1':      '#534AB7',
    'HMI2':      '#534AB7',
    'HMI3':      '#534AB7',
    'Gateway':   '#888780',
    'Attacker':  '#A32D2D',
    'C2 Server': '#854F0B',
}

# 4. LAYOUT
# Attacker on right, PLCs top-left, HMIs bottom, Gateway top-center
# Spread nodes more to reduce edge label overlap
pos = {
    '192.168.0.1':  (0.50, 0.92),  # Gateway — top center
    '192.168.0.11': (0.15, 0.62),  # PLC1 — left
    '192.168.0.12': (0.42, 0.62),  # PLC2 — center
    '192.168.0.41': (0.85, 0.62),  # Attacker — far right
    '192.168.0.21': (0.05, 0.25),  # HMI1 — bottom far left
    '192.168.0.22': (0.35, 0.20),  # HMI2 — bottom center-left
    '192.168.0.23': (0.62, 0.25),  # HMI3 — bottom center-right
    '192.168.0.42': (0.92, 0.25),  # C2 Server — bottom far right
}

# 5. DRAW
fig, ax = plt.subplots(figsize=(16, 10))
ax.set_facecolor('#F8F8F8')
fig.patch.set_facecolor('#F8F8F8')

# Nodes
node_color_list = [
    node_colors_map.get(role_map.get(ip, ''), '#888780')
    for ip in G_path.nodes()
]
nx.draw_networkx_nodes(
    G_path, pos,
    node_color=node_color_list,
    node_size=2400,
    alpha=0.95,
    ax=ax
)
node_labels = {
    ip: f"{role_map.get(ip, ip)}\n{ip}"
    for ip in G_path.nodes()
}
nx.draw_networkx_labels(
    G_path, pos,
    labels=node_labels,
    font_size=7.5,
    font_color='white',
    font_weight='bold',
    ax=ax
)

# Edges — separate arc per attack type to avoid overlap

arc_rads = {
    'mitm':      0.25,
    'ddos':      0.40,
    'replay':    0.12,
    'port-scan': 0.03,
}

for attack_type, color in attack_colors.items():
    edges_of_type = [
        (u, v) for u, v, d in G_path.edges(data=True)
        if d['attack_type'] == attack_type
    ]
    if not edges_of_type:
        continue

    nx.draw_networkx_edges(
        G_path, pos,
        edgelist=edges_of_type,
        edge_color=color,
        width=2.5,              # thicker lines
        alpha=0.85,
        arrows=True,
        arrowsize=25,           # bigger arrowhead
        arrowstyle='-|>',
        min_source_margin=30,   # gap at source node so arrow starts outside node
        min_target_margin=30,   # gap at target node so arrowhead visible
        connectionstyle=f'arc3,rad={arc_rads[attack_type]}',
        ax=ax
    )
#  6. EDGE LABELS (only on non-cluttered edges)
# Only label edges that go directly to/from Attacker to keep graph readable
attacker_ip = '192.168.0.41'
c2_ip       = '192.168.0.42'

edge_label_data = {}
for u, v, d in G_path.edges(data=True):
    if attacker_ip in (u, v) or c2_ip in (u, v):
        key = (u, v)
        atype = d['attack_type']
        edge_label_data.setdefault(key, {})
        edge_label_data[key][atype] = (
            edge_label_data[key].get(atype, 0) + d['count']
        )

edge_labels = {
    k: '\n'.join(f"{a}:{c}" for a, c in sorted(v.items()))
    for k, v in edge_label_data.items()
}
nx.draw_networkx_edge_labels(
    G_path, pos,
    edge_labels=edge_labels,
    font_size=6.5,
    font_color='#222222',
    bbox=dict(boxstyle='round,pad=0.25', fc='white', alpha=0.75, ec='none'),
    ax=ax
)

# 7. PHASE ANNOTATIONS (correctly placed, honestly labeled) ─
# NOTE: phases are analyst-inferred from first_seen timestamps,
# NOT labels present in the dataset.
annotations = [
    {
        'text':   'Step 1 (inferred)\nPort-scan · 12:15:42',
        'xy':     (0.85, 0.62),   # points to Attacker node
        'xytext': (0.72, 0.88),
        'color':  '#378ADD',
    },
    {
        'text':   'Step 2 (inferred)\nMITM · 12:17:06',
        'xy':     (0.285, 0.62),  # midpoint PLC1→PLC2 edge
        'xytext': (0.08, 0.80),
        'color':  '#D85A30',
    },
    {
        'text':   'Step 3 (inferred)\nReplay · 12:22:05',
        'xy':     (0.62, 0.25),   # points to HMI3
        'xytext': (0.45, 0.06),
        'color':  '#EF9F27',
    },
    {
        'text':   'Step 4 (inferred)\nDDoS · 12:34:42',
        'xy':     (0.64, 0.62),   # midpoint PLC2→Attacker edge
        'xytext': (0.58, 0.88),
        'color':  '#E24B4A',
    },
]
for ann in annotations:
    ax.annotate(
        ann['text'],
        xy=ann['xy'],
        xytext=ann['xytext'],
        xycoords='axes fraction',
        textcoords='axes fraction',
        fontsize=7.5,
        color=ann['color'],
        arrowprops=dict(arrowstyle='->', color=ann['color'], lw=1.0),
        bbox=dict(
            boxstyle='round,pad=0.3', fc='white',
            alpha=0.85, ec=ann['color'], lw=0.8
        )
    )

# 8. LEGENDS
attack_legend = [
    Line2D([0], [0], color=c, linewidth=2.5, label=a.capitalize())
    for a, c in attack_colors.items()
]
node_legend = [
    mpatches.Patch(color='#1D9E75', label='PLC'),
    mpatches.Patch(color='#534AB7', label='HMI'),
    mpatches.Patch(color='#A32D2D', label='Attacker'),
    mpatches.Patch(color='#854F0B', label='C2 Server'),
    mpatches.Patch(color='#888780', label='Gateway'),
]
leg1 = ax.legend(
    handles=attack_legend,
    title='Attack type',
    loc='upper left',
    fontsize=8,
    title_fontsize=9,
    framealpha=0.9
)
ax.add_artist(leg1)
ax.legend(
    handles=node_legend,
    title='Device role',
    loc='lower left',
    fontsize=8,
    title_fontsize=9,
    framealpha=0.9
)

# 9. TITLE
ax.set_title(
    'ICS-Flow Attack Flow Graph\n'
    'Node color = device role  ·  Edge color = attack type  ·  '
    'Edge labels = flow count (attacker edges only)\n'
    'Steps are analyst-inferred from first_seen timestamps — '
    'not dataset labels',
    fontsize=10,
    fontweight='bold',
    pad=14
)

ax.axis('off')
plt.tight_layout()
plt.savefig('week3_attack_flow_graph.png', dpi=150, bbox_inches='tight')
plt.show()
print("Saved: week3_attack_flow_graph.png")
````

## **BATADAL Dataset**

````python
# WEEK 3 — BATADAL Dataset Graph Construction
# Dataset: Battle of the Attack Detection Algorithms (BATADAL)
# Source: http://www.batadal.net/data.html
# File: BATADAL_dataset04.csv (training with attacks)


import pandas as pd
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.lines import Line2D
from scipy.stats import pearsonr
import warnings
warnings.filterwarnings('ignore')

# 1. LOAD DATA
df = pd.read_csv('BATADAL_dataset04.csv')

# Clean column name
df.columns = df.columns.str.strip()

print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nFirst rows:")
print(df.head(3))
print("\nAttack label distribution:")
print(df['ATT_FLAG'].value_counts())
````

````python
# Fix 1: ATT_FLAG uses -999 for normal, 1 for attack (not 0/1)
df['ATT_FLAG'] = df['ATT_FLAG'].replace(-999, 0)

print("Fixed label distribution:")
print(df['ATT_FLAG'].value_counts())

# So sensor type detection needs updating
sensor_cols = [c for c in df.columns if c not in ['DATETIME', 'ATT_FLAG']]

def get_sensor_type(col):
    if col.startswith('L_T'):   return 'Tank'
    if col.startswith('F_PU'):  return 'Pump_Flow'
    if col.startswith('F_V'):   return 'Valve_Flow'
    if col.startswith('P_J'):   return 'Pressure'
    if col.startswith('S_PU'):  return 'Pump_Status'
    if col.startswith('S_V'):   return 'Valve_Status'
    return 'Unknown'

sensor_type_map = {col: get_sensor_type(col) for col in sensor_cols}

print("\n=== Sensor inventory ===")
for stype in ['Tank', 'Pump_Flow', 'Valve_Flow',
              'Pressure', 'Pump_Status', 'Valve_Status']:
    members = [c for c, t in sensor_type_map.items() if t == stype]
    print(f"{stype:15s}: {len(members):3d}  →  {members}")

# Fix 3: Split normal vs attack with corrected label
df_normal = df[df['ATT_FLAG'] == 0][sensor_cols]
df_attack = df[df['ATT_FLAG'] == 1][sensor_cols]

print(f"\nNormal timesteps : {len(df_normal)}")
print(f"Attack timesteps : {len(df_attack)}")
print(f"Attack ratio     : {len(df_attack)/len(df)*100:.1f}%")
````

### **BUILD CORRELATION GRAPHS**

````python
#  4. BUILD CORRELATION GRAPHS
from scipy.stats import pearsonr
import warnings
warnings.filterwarnings('ignore')

CORRELATION_THRESHOLD = 0.7

def build_correlation_graph(data, threshold=0.7):
    G = nx.Graph()
    for col in sensor_cols:
        G.add_node(col, sensor_type=sensor_type_map[col])

    cols = data.columns.tolist()
    for i in range(len(cols)):
        for j in range(i+1, len(cols)):
            col_i, col_j = cols[i], cols[j]
            if data[col_i].std() == 0 or data[col_j].std() == 0:
                continue
            corr, pval = pearsonr(
                data[col_i].fillna(0),
                data[col_j].fillna(0)
            )
            if abs(corr) >= threshold and pval < 0.05:
                G.add_edge(col_i, col_j,
                           weight=round(abs(corr), 3),
                           correlation=round(corr, 3))
    return G

print("Building normal graph...")
G_normal = build_correlation_graph(df_normal, CORRELATION_THRESHOLD)

print("Building attack graph...")
G_attack = build_correlation_graph(df_attack, CORRELATION_THRESHOLD)

print(f"\nNormal graph → nodes: {G_normal.number_of_nodes()}, edges: {G_normal.number_of_edges()}")
print(f"Attack graph → nodes: {G_attack.number_of_nodes()}, edges: {G_attack.number_of_edges()}")
````

### **ANOMALY ANALYSIS**

````python
# 5. ANOMALY ANALYSIS
normal_edges = set(G_normal.edges())
attack_edges = set(G_attack.edges())

new_in_attack  = attack_edges - normal_edges
lost_in_attack = normal_edges - attack_edges

print(f"New edges during attack  : {len(new_in_attack)}")
print(f"Lost edges during attack : {len(lost_in_attack)}")
print(f"Stable edges (both)      : {len(normal_edges & attack_edges)}")

print("\n=== New anomalous edges (appeared only during attack) ===")
for e in sorted(new_in_attack):
    w = G_attack[e[0]][e[1]]['weight']
    t0 = sensor_type_map[e[0]]
    t1 = sensor_type_map[e[1]]
    print(f"  {e[0]:10s} ({t0:12s}) ↔ {e[1]:10s} ({t1:12s})  corr={w}")

# Anomalous sensors by value deviation
def detect_anomalous_sensors(df_normal, df_attack, threshold_std=2.0):
    anomalous = {}
    for col in sensor_cols:
        normal_mean = df_normal[col].mean()
        normal_std  = df_normal[col].std()
        attack_mean = df_attack[col].mean()
        if normal_std == 0:
            continue
        deviation = abs(attack_mean - normal_mean) / normal_std
        if deviation >= threshold_std:
            anomalous[col] = round(deviation, 2)
    return anomalous

anomalous_sensors = detect_anomalous_sensors(df_normal, df_attack)

print(f"\n=== Anomalous sensors during attack ({len(anomalous_sensors)} found) ===")
for sensor, dev in sorted(anomalous_sensors.items(), key=lambda x: -x[1]):
    stype = sensor_type_map[sensor]
    print(f"  {sensor:12s} ({stype:12s})  deviation = {dev} std")
````

````python
#  6. VISUALIZATION
type_colors = {
    'Tank':         '#1D9E75',
    'Pump_Flow':    '#378ADD',
    'Valve_Flow':   '#534AB7',
    'Pressure':     '#EF9F27',
    'Pump_Status':  '#E24B4A',
    'Valve_Status': '#888780',
    'Unknown':      '#333333',
}

def draw_batadal_graph(G, anomalous, title, ax):
    if G.number_of_nodes() == 0:
        ax.text(0.5, 0.5, 'No graph', ha='center')
        return

    # Group nodes by type for a cleaner layout
    # Place same sensor types near each other
    pos = nx.spring_layout(G, seed=42, k=2.0)

    node_colors     = [type_colors.get(G.nodes[n].get('sensor_type','Unknown'), '#888780') for n in G.nodes()]
    node_edge_colors= ['#FF0000' if n in anomalous else '#ffffff' for n in G.nodes()]
    node_sizes      = [1000 if n in anomalous else 600 for n in G.nodes()]
    edge_weights    = [G[u][v]['weight'] * 3 for u, v in G.edges()]

    nx.draw_networkx_edges(G, pos,
        width=edge_weights, alpha=0.35,
        edge_color='#AAAAAA', ax=ax)

    nx.draw_networkx_nodes(G, pos,
        node_color=node_colors,
        node_size=node_sizes,
        edgecolors=node_edge_colors,
        linewidths=2.5,
        alpha=0.95, ax=ax)

    nx.draw_networkx_labels(G, pos,
        font_size=6.5,
        font_color='white',
        font_weight='bold', ax=ax)

    # Annotate anomalous nodes
    for node, dev in anomalous.items():
        if node in pos:
            x, y = pos[node]
            ax.annotate(f'Δ{dev}σ',
                xy=(x, y), xytext=(x+0.1, y+0.1),
                fontsize=7, color='red', fontweight='bold',
                arrowprops=dict(arrowstyle='->', color='red', lw=0.8))

    ax.set_title(title, fontsize=10, fontweight='bold', pad=10)
    ax.axis('off')

# Draw
fig, axes = plt.subplots(1, 2, figsize=(20, 9))
fig.patch.set_facecolor('#F8F8F8')

draw_batadal_graph(
    G_normal, {},
    f'Normal Operation\n'
    f'{G_normal.number_of_nodes()} sensors · '
    f'{G_normal.number_of_edges()} correlated pairs · '
    f'threshold={CORRELATION_THRESHOLD}',
    axes[0]
)
draw_batadal_graph(
    G_attack, anomalous_sensors,
    f'Attack Period\n'
    f'{G_attack.number_of_nodes()} sensors · '
    f'{G_attack.number_of_edges()} correlated pairs · '
    f'Red border = anomalous sensor',
    axes[1]
)

# Legend
legend_handles = [
    mpatches.Patch(color=c, label=t.replace('_',' '))
    for t, c in type_colors.items() if t != 'Unknown'
]
legend_handles.append(
    mpatches.Patch(facecolor='white', edgecolor='red',
                   linewidth=2.5, label='Anomalous sensor')
)
fig.legend(handles=legend_handles,
           title='Sensor type',
           loc='lower center', ncol=7,
           fontsize=8, title_fontsize=9,
           framealpha=0.9, bbox_to_anchor=(0.5, -0.02))

plt.suptitle(
    'BATADAL Water Distribution System — Sensor Correlation Graph\n'
    'Edges = Pearson correlation ≥ 0.7  ·  '
    'Node size = larger if anomalous during attack  ·  '
    'Δσ = deviation from normal mean',
    fontsize=11, fontweight='bold', y=1.01
)
plt.tight_layout()
plt.savefig('week3_batadal_graph.png', dpi=150, bbox_inches='tight')
plt.show()
print("Saved: week3_batadal_graph.png")
````

````python
# 1. SETUP
import pandas as pd
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.lines import Line2D
from collections import Counter

print(f"ICS-Flow records : {len(df_graph)}")
print(f"Attack types     : {df_graph['NST_M_Label'].value_counts().to_dict()}")
````

### 2. Device role — single source of truth

This is the **same `role_map`** already defined in the Week 3 *Attack Flow Graph*
cell. It's testbed configuration (which physical device each IP is).

````python
# 2. DEVICE ROLE (identical to the Week 3 role_map — single source of truth)
role_map = {
    '192.168.0.1':  'Gateway',
    '192.168.0.11': 'PLC1',
    '192.168.0.12': 'PLC2',
    '192.168.0.21': 'HMI1',
    '192.168.0.22': 'HMI2',
    '192.168.0.23': 'HMI3',
    '192.168.0.41': 'Attacker',
    '192.168.0.42': 'C2 Server',
}

def node_type_of(ip: str) -> str:
    """Asset vs Attacker, derived from role_map."""
    role = role_map.get(ip, 'Unknown')
    return 'Attacker' if role in ('Attacker', 'C2 Server') else 'Asset'

print("Device roles:")
for ip, role in role_map.items():
    print(f"  {ip:15s} -> {role:10s} ({node_type_of(ip)})")
````

### 3. Flow-derived device statistics

`flow_count` and `source_ratio` computed directly from `df_graph` — same
numbers as the Week 2 `role_df` table, just wrapped in a reusable function so
Week 4 doesn't re-type them.

````python
# 3. FLOW-DERIVED STATS
def build_device_stats(df_graph: pd.DataFrame) -> pd.DataFrame:
    src = df_graph['sIPs'].value_counts().rename('as_source')
    dst = df_graph['rIPs'].value_counts().rename('as_destination')
    stats = pd.concat([src, dst], axis=1).fillna(0).astype(int)
    stats['flow_count'] = stats['as_source'] + stats['as_destination']
    stats['source_ratio'] = (stats['as_source'] / stats['flow_count']).round(3)
    return stats

device_stats = build_device_stats(df_graph)
device_stats
````

### 4. Threat-intel enrichment: MITRE ATT&CK techniques + CVEs

These **cannot** be derived from network flow data — a CSV of packet flows has
no concept of a CVE ID or a MITRE technique ID. This is standard practice for
real ICS knowledge graphs: topology + behavior comes from telemetry, vulnerability
context comes from a threat-intel feed. They're merged, not conflated — only the
`maps_to` field ties a technique back to the dataset's own `NST_M_Label` values.

````python
# 4. MITRE ATT&CK ICS TECHNIQUE NODES (external reference, mapped to dataset labels)
technique_nodes = {
    'T0830': {'node_type': 'Technique', 'name': 'Adversary-in-the-Middle',
              'tactic': 'Collection, Lateral Movement', 'maps_to': 'mitm',
              'description': 'Attacker intercepts communications between ICS devices',
              'severity': 'High'},
    'T0814': {'node_type': 'Technique', 'name': 'Denial of Service',
              'tactic': 'Inhibit Response Function', 'maps_to': 'ddos',
              'description': 'Flood ICS devices to prevent legitimate communication',
              'severity': 'High'},
    'T0843': {'node_type': 'Technique', 'name': 'Program Download / Replay',
              'tactic': 'Impair Process Control', 'maps_to': 'replay',
              'description': 'Replay captured ICS packets to manipulate physical process',
              'severity': 'Critical'},
    'T0846': {'node_type': 'Technique', 'name': 'Remote System Discovery',
              'tactic': 'Discovery', 'maps_to': 'port-scan',
              'description': 'Scan ICS network to identify devices and open ports',
              'severity': 'Medium'},
}
label_to_technique = {v['maps_to']: k for k, v in technique_nodes.items()}

# CVE NODES (external reference — real CVEs relevant to Modbus/TCP ICS)
cve_nodes = {
    'CVE-2019-13945': {'node_type': 'CVE', 'cvss_score': 6.8,
                        'description': 'Siemens S7 PLC — unauthenticated Modbus access allows replay',
                        'affects': ['192.168.0.11', '192.168.0.12'], 'enables': ['T0843', 'T0830']},
    'CVE-2020-15782': {'node_type': 'CVE', 'cvss_score': 9.8,
                        'description': 'Siemens PLC memory protection bypass — enables remote code exec',
                        'affects': ['192.168.0.11', '192.168.0.12'], 'enables': ['T0843']},
    'CVE-2019-10953': {'node_type': 'CVE', 'cvss_score': 7.5,
                        'description': 'ABB/Siemens HMI — denial of service via malformed packets',
                        'affects': ['192.168.0.21', '192.168.0.22', '192.168.0.23'], 'enables': ['T0814']},
    'CVE-2015-5374': {'node_type': 'CVE', 'cvss_score': 7.8,
                       'description': 'Siemens SCALANCE — ARP spoofing enables MITM on ICS network',
                       'affects': ['192.168.0.11', '192.168.0.12', '192.168.0.21', '192.168.0.23'],
                       'enables': ['T0830']},
}

print(f"Technique nodes: {len(technique_nodes)}  |  CVE nodes: {len(cve_nodes)}")
````

### 5. Build the Knowledge Graph


````python
# 5. BUILD KNOWLEDGE GRAPH
def build_knowledge_graph(df_graph: pd.DataFrame) -> nx.MultiDiGraph:
    stats = build_device_stats(df_graph)
    KG = nx.MultiDiGraph()

    # Asset / Attacker nodes: role from role_map, stats from data
    for ip, role in role_map.items():
        row = stats.loc[ip] if ip in stats.index else None
        KG.add_node(
            ip,
            node_type=node_type_of(ip),
            role=role,
            label=role,
            flow_count=int(row['flow_count']) if row is not None else 0,
            source_ratio=float(row['source_ratio']) if row is not None else 0.0,
        )

    #  Technique nodes
    for tid, attrs in technique_nodes.items():
        KG.add_node(tid, technique_id=tid, **attrs)

    # CVE nodes
    for cve_id, attrs in cve_nodes.items():
        KG.add_node(cve_id, cve_id=cve_id, **attrs)

    # COMMUNICATES_WITH: every real attack flow, grouped
    attack_timeline = (
        df_graph[df_graph['NST_M_Label'] != 'Normal']
        .groupby(['sIPs', 'rIPs', 'NST_M_Label'])
        .agg(count=('NST_M_Label', 'size'), first_seen=('startDate', 'min'))
        .reset_index()
    )
    for _, r in attack_timeline.iterrows():
        KG.add_edge(r['sIPs'], r['rIPs'], edge_type='COMMUNICATES_WITH',
                    attack_type=r['NST_M_Label'], count=int(r['count']),
                    first_seen=str(r['first_seen']))

    #  USES_TECHNIQUE: derived from who actually SOURCES each attack label
    attack_sources = (
        df_graph[df_graph['NST_M_Label'] != 'Normal']
        .groupby(['sIPs', 'NST_M_Label']).size()
        .reset_index(name='count')
    )
    for _, r in attack_sources.iterrows():
        label = r['NST_M_Label']
        if label not in label_to_technique:
            continue
        KG.add_edge(r['sIPs'], label_to_technique[label],
                    edge_type='USES_TECHNIQUE', weight=int(r['count']))

    # TARGETS: derived from who actually RECEIVES each attack label
    attack_targets = (
        df_graph[df_graph['NST_M_Label'] != 'Normal']
        .groupby(['rIPs', 'NST_M_Label']).size()
        .reset_index(name='count')
    )
    for _, r in attack_targets.iterrows():
        label = r['NST_M_Label']
        if label not in label_to_technique:
            continue
        KG.add_edge(label_to_technique[label], r['rIPs'],
                    edge_type='TARGETS', weight=int(r['count']))

    # VULNERABLE_TO / ENABLES: from the CVE feed (external, unchanged)
    for cve_id, cve in cve_nodes.items():
        for ip in cve['affects']:
            # Asset → VULNERABLE_TO → CVE
           KG.add_edge(ip, cve_id, edge_type='VULNERABLE_TO', cvss=cve['cvss_score'])
        for tid in cve['enables']:
            # CVE → ENABLES → Technique
            KG.add_edge(cve_id, tid, edge_type='ENABLES')

    return KG

KG = build_knowledge_graph(df_graph)

print("=== Knowledge Graph Built ===")
print(f"Total nodes : {KG.number_of_nodes()}")
print(f"Total edges : {KG.number_of_edges()}")
print("\nNode type breakdown:")
for ntype, count in Counter(d['node_type'] for _, d in KG.nodes(data=True)).items():
    print(f"  {ntype:15s}: {count}")
print("\nEdge type breakdown:")
for etype, count in Counter(d['edge_type'] for _, _, d in KG.edges(data=True)).items():
    print(f"  {etype:25s}: {count}")
````

### 6. Sanity check: does USES_TECHNIQUE / TARGETS now match the real attack direction?


````python
#  6. SANITY CHECK — confirm edges now reflect real flow direction
print("USES_TECHNIQUE edges (source_ip -> technique, weight = real flow count):")
for u, v, d in KG.edges(data=True):
    if d.get('edge_type') == 'USES_TECHNIQUE':
        print(f"  {u:15s} -> {v}  (weight={d['weight']})")

print("\nTARGETS edges (technique -> target_ip, weight = real flow count):")
for u, v, d in KG.edges(data=True):
    if d.get('edge_type') == 'TARGETS':
        print(f"  {u} -> {v:15s}  (weight={d['weight']})")
````

### 7. Node risk scores

 Weighted formula(`0.5 * attack_score + 0.2 * flow_score + 0.3 * cve_score` for assets)

````python
# 7. COMPUTE NODE RISK SCORES
def compute_risk_scores(KG: nx.MultiDiGraph) -> dict:
    severity_weight = {'Critical': 1.0, 'High': 0.75, 'Medium': 0.5, 'Low': 0.25}
    risk_scores = {}

    for node, data in KG.nodes(data=True):
        ntype = data.get('node_type', '')

        if ntype == 'Asset':
            attack_labels_seen = {
                d.get('attack_type') for _, _, d in KG.edges(node, data=True)
                if d.get('edge_type') == 'COMMUNICATES_WITH'
            } - {None}
            attack_score = min(len(attack_labels_seen) / 4.0, 1.0)
            flow_score = min(data.get('flow_count', 0) / 30000, 1.0)
            cve_score = max(
                [c['cvss_score'] / 10.0 for c in cve_nodes.values() if node in c['affects']],
                default=0.0)
            risk = round(attack_score * 0.5 + flow_score * 0.2 + cve_score * 0.3, 3)

        elif ntype == 'Technique':
            sev = data.get('severity', 'Medium')
            n_targets = sum(1 for _, _, d in KG.out_edges(node, data=True)
                             if d.get('edge_type') == 'TARGETS')
            risk = round(severity_weight[sev] * 0.7 + min(n_targets / 6, 1.0) * 0.3, 3)

        elif ntype == 'CVE':
            risk = round(data['cvss_score'] / 10.0, 3)

        elif ntype == 'Attacker':
            risk = 1.0

        else:
            risk = 0.0

        risk_scores[node] = risk
        KG.nodes[node]['risk_score'] = risk

    return risk_scores

risk_scores = compute_risk_scores(KG)

print("=== Node Risk Scores ===")
for node, score in sorted(risk_scores.items(), key=lambda x: -x[1]):
    ntype = KG.nodes[node].get('node_type', '')
    label = KG.nodes[node].get('label', node)
    print(f"  {label:20s} ({ntype:10s})  risk = {score:.3f}")
````

## **Part B — PyTorch Geometric GNN Pipeline**

**Caveat before training:** the ICS-Flow testbed has only **9 devices**. A
*node* classifier trained on 9 examples can't be split into a meaningful
train/test set. So alongside the node classifier,
this section also builds an **edge/flow-level classifier** — each of the
~44k individual flows becomes one training example (real train/test split,
precision/recall/F1) — which is the statistically sound option at this
dataset's scale, and the one that plugs into BATADAL cross-validation in
Week 5 without changes.

### 12. Node features and topology → PyG `Data` object



````python
import torch
import torch.nn.functional as F
from torch_geometric.data import Data
from torch_geometric.nn import GATConv

def build_node_features(df_graph: pd.DataFrame):
    """One feature row per device IP, computed purely from flow stats.
    Restricted to known devices (role_map) — filters out non-asset
    addresses like multicast (224.0.0.251, mDNS) that appear in sIPs/
    rIPs but aren't part of the ICS asset inventory."""
    ips = sorted(set(role_map.keys()) &
                 (set(df_graph['sIPs']) | set(df_graph['rIPs'])))
    ip_index = {ip: i for i, ip in enumerate(ips)}

    src = df_graph['sIPs'].value_counts()
    dst = df_graph['rIPs'].value_counts()
    attack_df = df_graph[df_graph['NST_M_Label'] != 'Normal']
    attack_src = attack_df['sIPs'].value_counts()
    attack_dst = attack_df['rIPs'].value_counts()
    proto_div = df_graph.groupby('sIPs')['protocol'].nunique()

    rows = []
    for ip in ips:
        as_src, as_dst = src.get(ip, 0), dst.get(ip, 0)
        total = as_src + as_dst
        rows.append([
            as_src, as_dst,
            as_src / total if total else 0.0,             # source_ratio
            attack_src.get(ip, 0) / max(as_src, 1),        # frac of outgoing flows that are attacks
            attack_dst.get(ip, 0) / max(as_dst, 1),         # frac of incoming flows that are attacks
            proto_div.get(ip, 0),
        ])
    return ips, np.array(rows, dtype=np.float32)

def build_edge_index(df_graph: pd.DataFrame, ip_index: dict) -> torch.Tensor:
    """Communication topology restricted to known devices — drops any
    flow touching a non-asset address (e.g. multicast) since it has no
    node to attach to in ip_index."""
    pairs = df_graph[['sIPs', 'rIPs']].drop_duplicates()
    pairs = pairs[pairs['sIPs'].isin(ip_index) & pairs['rIPs'].isin(ip_index)]
    src = pairs['sIPs'].map(ip_index).to_numpy()
    dst = pairs['rIPs'].map(ip_index).to_numpy()
    return torch.tensor(np.vstack([src, dst]), dtype=torch.long)

def build_pyg_data(df_graph: pd.DataFrame) -> Data:
    ips, feats = build_node_features(df_graph)
    ip_index = {ip: i for i, ip in enumerate(ips)}
    edge_index = build_edge_index(df_graph, ip_index)
    data = Data(x=torch.tensor(feats, dtype=torch.float), edge_index=edge_index)
    data.ip_list = ips
    data.ip_index = ip_index
    return data

pyg_data = build_pyg_data(df_graph)
print(pyg_data)
print("Node feature matrix shape:", pyg_data.x.shape)

# Add node labels and masks — needed for node classifier
node_labels_map = {
    ip: 1 if node_type_of(ip) == 'Attacker' else 0
    for ip in pyg_data.ip_list
}
pyg_data.y = torch.tensor(
    [node_labels_map.get(ip, 0) for ip in pyg_data.ip_list],
    dtype=torch.long
)

# Train/test masks — hold out HMI2 and Gateway as test
n = len(pyg_data.ip_list)
train_mask = torch.ones(n, dtype=torch.bool)
test_mask  = torch.zeros(n, dtype=torch.bool)
test_ips   = {'192.168.0.22', '192.168.0.1'}
for i, ip in enumerate(pyg_data.ip_list):
    if ip in test_ips:
        train_mask[i] = False
        test_mask[i]  = True

pyg_data.train_mask = train_mask
pyg_data.test_mask  = test_mask

print(f"\nNode labels assigned:")
for ip, label in zip(pyg_data.ip_list, pyg_data.y.tolist()):
    print(f"  {ip:15s} ({role_map.get(ip):12s}) label={label}")
print(f"\nTrain nodes : {train_mask.sum().item()}")
print(f"Test nodes  : {test_mask.sum().item()}")
````

### 13. GraphSAGE model definitions

````python
from torch_geometric.nn import SAGEConv

class SAGEEmbedder(torch.nn.Module):
    """Node embedder using GraphSAGE aggregation.
    Chosen over GAT because:
    - Inductive: generalises to unseen BATADAL nodes in Week 5
    Reference: Hamilton et al., NeurIPS 2017, arxiv.org/abs/1706.02216
    """
    def __init__(self, in_dim, hidden_dim=16, out_dim=16):
        super().__init__()
        self.conv1 = SAGEConv(in_dim, hidden_dim)
        self.conv2 = SAGEConv(hidden_dim, out_dim)
        self.dropout = torch.nn.Dropout(p=0.3)

    def forward(self, x, edge_index):
        x = F.relu(self.conv1(x, edge_index))
        x = self.dropout(x)
        x = self.conv2(x, edge_index)
        return x

    def encode(self, x, edge_index):
        """Return embeddings for Week 5 BATADAL transfer."""
        return self.forward(x, edge_index)


class NodeClassifier(torch.nn.Module):
    """Node-level risk classifier using GraphSAGE embeddings.
    NOTE: 9 nodes — no meaningful test split possible.
    Reported with this caveat as supervisor requested."""
    def __init__(self, in_dim, hidden_dim=16):
        super().__init__()
        self.embedder = SAGEEmbedder(in_dim, hidden_dim, out_dim=hidden_dim)
        self.classifier = torch.nn.Linear(hidden_dim, 2)

    def forward(self, x, edge_index):
        return self.classifier(self.embedder(x, edge_index))


class EdgeClassifier(torch.nn.Module):
    """Flow-level classifier — 44k examples, real train/test split.
    GraphSAGE embeddings concatenated per edge endpoint → MLP head."""
    def __init__(self, in_dim, hidden_dim=16):
        super().__init__()
        self.embedder = SAGEEmbedder(in_dim, hidden_dim, out_dim=hidden_dim)
        self.head = torch.nn.Sequential(
            torch.nn.Linear(hidden_dim * 2, hidden_dim),
            torch.nn.ReLU(),
            torch.nn.Linear(hidden_dim, 2),
        )

    def forward(self, x, edge_index, query_edges):
        h = self.embedder(x, edge_index)
        return self.head(
            torch.cat([h[query_edges[0]], h[query_edges[1]]], dim=1)
        )

print("Model classes defined: SAGEEmbedder, NodeClassifier, EdgeClassifier")
````

### 14. Node-level training

Labels each of the 9 devices as anomalous (1) if it's the `Attacker`/`C2
Server`, else normal (0). Trains and reports accuracy — but with no held-out
test set, since 9 examples can't be split meaningfully.

````python
# 14. NODE-LEVEL TRAINING
def train_node_classifier(data: Data, node_labels: dict, epochs=100):
    torch.manual_seed(42)
    y = torch.tensor([node_labels.get(ip, 0) for ip in data.ip_list], dtype=torch.long)
    model = NodeClassifier(in_dim=data.x.shape[1])
    optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=5e-4)

    model.train()
    for epoch in range(epochs):
        optimizer.zero_grad()
        loss = F.cross_entropy(model(data.x, data.edge_index), y)
        loss.backward()
        optimizer.step()

    model.eval()
    with torch.no_grad():
        pred = model(data.x, data.edge_index).argmax(dim=1)
        acc = (pred == y).float().mean().item()

    return model, {'train_accuracy': acc, 'n_nodes': len(data.ip_list),
                    'note': 'No held-out test set — 9 nodes is too few to split meaningfully.'}

# Anomalous = Attacker / C2 Server role (from role_map, not re-derived)
node_labels = {ip: 1 if node_type_of(ip) == 'Attacker' else 0 for ip in role_map}
node_model, node_metrics = train_node_classifier(pyg_data, node_labels)
print("Node classifier:", node_metrics)
````

### 13. Edge/flow-level training

Each individual flow row is its own example — real train/test split with
precision/recall/F1, not just accuracy (attack traffic is the minority
class, so accuracy alone would be misleading).


````python
# ── 13. EDGE-LEVEL TRAINING (flow-level examples, real train/test split) ──
def build_edge_labels(df_graph: pd.DataFrame, ip_index: dict):
    df = df_graph[df_graph['sIPs'].isin(ip_index) & df_graph['rIPs'].isin(ip_index)]
    src = torch.tensor(df['sIPs'].map(ip_index).to_numpy(), dtype=torch.long)
    dst = torch.tensor(df['rIPs'].map(ip_index).to_numpy(), dtype=torch.long)
    y = torch.tensor((df['NST_M_Label'] != 'Normal').astype(int).to_numpy(), dtype=torch.long)
    return torch.stack([src, dst]), y

def train_edge_classifier(data: Data, df_graph: pd.DataFrame, epochs=200,
                           test_ratio=0.2, seed=42):
    torch.manual_seed(seed)
    query_edges, y = build_edge_labels(df_graph, data.ip_index)

    n = query_edges.shape[1]
    g = torch.Generator().manual_seed(seed)
    perm = torch.randperm(n, generator=g)
    n_test = max(1, int(n * test_ratio))
    test_idx, train_idx = perm[:n_test], perm[n_test:]

    # Class weighting — without this the model just predicts the majority
    # class (~83% Normal) and looks "accurate" while learning nothing.
    # Weight inversely proportional to class frequency in the TRAIN split.
    train_y = y[train_idx]
    class_counts = torch.bincount(train_y, minlength=2).float()
    class_weights = train_y.shape[0] / (2.0 * class_counts.clamp(min=1))
    print(f"Class counts (train): normal={int(class_counts[0])}, "
          f"anomalous={int(class_counts[1])}  ->  weights={class_weights.tolist()}")

    model = EdgeClassifier(in_dim=data.x.shape[1])
    optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=5e-4)

    model.train()
    for epoch in range(epochs):
        optimizer.zero_grad()
        out = model(data.x, data.edge_index, query_edges[:, train_idx])
        loss = F.cross_entropy(out, y[train_idx], weight=class_weights)
        loss.backward()
        optimizer.step()

    model.eval()
    with torch.no_grad():
        test_pred = model(data.x, data.edge_index, query_edges[:, test_idx]).argmax(dim=1)
        acc = (test_pred == y[test_idx]).float().mean().item()
        tp = ((test_pred == 1) & (y[test_idx] == 1)).sum().item()
        fp = ((test_pred == 1) & (y[test_idx] == 0)).sum().item()
        fn = ((test_pred == 0) & (y[test_idx] == 1)).sum().item()
        tn = ((test_pred == 0) & (y[test_idx] == 0)).sum().item()
        precision = tp / (tp + fp) if (tp + fp) else 0.0
        recall = tp / (tp + fn) if (tp + fn) else 0.0
        f1 = 2 * precision * recall / (precision + recall) if (precision + recall) else 0.0

    class_counts_test = torch.bincount(y[test_idx], minlength=2).tolist()
    return model, {'test_accuracy': acc, 'precision': precision, 'recall': recall, 'f1': f1,
                    'n_train_edges': len(train_idx), 'n_test_edges': len(test_idx),
                    'train_class_counts': {'normal': int(class_counts[0]), 'anomalous': int(class_counts[1])},
                    'test_class_counts': {'normal': class_counts_test[0], 'anomalous': class_counts_test[1]},
                    'confusion_matrix': {'tp': tp, 'fp': fp, 'fn': fn, 'tn': tn}}

edge_model, edge_metrics = train_edge_classifier(pyg_data, df_graph)
print("Edge classifier:", edge_metrics)
````

## **Part D — BATADAL Cross-Dataset Validation (windowed)**

- BATADAL: 4,177 hourly rows / 174 days, 219 attack-flagged hours (~7 events)
- Sliding 24h window, 6h stride -> 693 windows, labelled anomalous if it contains any attack hour (56 positive / 693, ~8%)

- Topology fixed from G_normal (Week 3 normal-period correlation graph) since 24 points/window is too few to re-estimate correlation reliably. Node features vary per window instead: mean, std, and z-score deviation from each sensor's own normal-period baseline.

- Stratified 5-fold CV (not a single split) since positives are only ~8%.

````python
# Build sliding-window graph instances
# Reuses df, sensor_cols, sensor_type_map, df_normal, G_normal already built
# in the Week 3 BATADAL section above.

import torch
import numpy as np
import networkx as nx
from torch_geometric.data import Data

WINDOW_SIZE = 24   # hours
STRIDE = 6         # hours

sensor_list = sorted(sensor_cols)
sensor_index = {s: i for i, s in enumerate(sensor_list)}
type_codes = {t: i for i, t in enumerate(sorted(set(sensor_type_map.values())))}

#The real sensor correlation structure from the full normal period
# (built in Week 3). Using this instead of a per-window correlation graph avoids
# estimating correlation from only 24 data points, which is too noisy to trust.
fixed_edges = [(sensor_index[u], sensor_index[v]) for u, v in G_normal.edges()
               if u in sensor_index and v in sensor_index]
# Make undirected message passing explicit (both directions)
fixed_edges = fixed_edges + [(v, u) for u, v in fixed_edges]
edge_index = torch.tensor(fixed_edges, dtype=torch.long).T if fixed_edges else torch.empty((2, 0), dtype=torch.long)
print(f"Topology: {len(sensor_list)} sensors, {edge_index.shape[1]} directed edges (from G_normal)")

# Per-sensor normal-period baseline for z-score features
baseline_mean = df_normal[sensor_list].mean()
baseline_std = df_normal[sensor_list].std().replace(0, 1e-9)

is_attack = (df['ATT_FLAG'] == 1).astype(int).to_numpy()
n_rows = len(df)
starts = list(range(0, n_rows - WINDOW_SIZE + 1, STRIDE))

window_graphs, window_labels = [], []
for s in starts:
    window = df[sensor_list].iloc[s:s + WINDOW_SIZE]
    label = int(is_attack[s:s + WINDOW_SIZE].sum() > 0)

    w_mean = window.mean()
    w_std = window.std().fillna(0)
    zscore = ((w_mean - baseline_mean) / baseline_std).clip(-5, 5)  # deviation from own baseline

    feats = []
    for sensor in sensor_list:
        vmin, vmax = df[sensor].min(), df[sensor].max()
        mean_norm = (w_mean[sensor] - vmin) / (vmax - vmin + 1e-9)
        std_norm = w_std[sensor] / (vmax - vmin + 1e-9)
        feats.append([mean_norm, std_norm, zscore[sensor],
                      type_codes[sensor_type_map[sensor]]])

    x = torch.tensor(np.array(feats, dtype=np.float32))
    g = Data(x=x, edge_index=edge_index, y=torch.tensor([label], dtype=torch.long))
    window_graphs.append(g)
    window_labels.append(label)

window_labels = np.array(window_labels)
print(f"\nTotal windows   : {len(window_graphs)}")
print(f"Positive windows: {window_labels.sum()}  ({window_labels.mean()*100:.1f}%)")
print(f"Negative windows: {(window_labels == 0).sum()}")
````

````python
#  Graph-level GraphSAGE classifier
import torch.nn.functional as F
from torch_geometric.nn import SAGEConv, global_mean_pool
from torch_geometric.loader import DataLoader
from sklearn.model_selection import StratifiedKFold

class WindowGraphClassifier(torch.nn.Module):
    """GraphSAGE embedder + mean pooling + linear head -> per-window
    (graph-level) anomaly classification, mirroring the SAGEEmbedder
    architecture used for ICS-Flow, adapted from edge-level to graph-level."""
    def __init__(self, in_dim, hidden_dim=16):
        super().__init__()
        self.conv1 = SAGEConv(in_dim, hidden_dim)
        self.conv2 = SAGEConv(hidden_dim, hidden_dim)
        self.dropout = torch.nn.Dropout(p=0.3)
        self.head = torch.nn.Linear(hidden_dim, 2)

    def forward(self, x, edge_index, batch):
        h = F.relu(self.conv1(x, edge_index))
        h = self.dropout(h)
        h = self.conv2(h, edge_index)
        h = global_mean_pool(h, batch)
        return self.head(h)


def run_fold(train_graphs, test_graphs, epochs=100, seed=42):
    torch.manual_seed(seed)
    train_loader = DataLoader(train_graphs, batch_size=16, shuffle=True)
    test_loader = DataLoader(test_graphs, batch_size=32, shuffle=False)

    train_y = torch.cat([g.y for g in train_graphs])
    class_counts = torch.bincount(train_y, minlength=2).float()
    class_weights = train_y.shape[0] / (2.0 * class_counts.clamp(min=1))

    model = WindowGraphClassifier(in_dim=train_graphs[0].x.shape[1])
    optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=5e-4)

    model.train()
    for epoch in range(epochs):
        for batch in train_loader:
            optimizer.zero_grad()
            out = model(batch.x, batch.edge_index, batch.batch)
            loss = F.cross_entropy(out, batch.y, weight=class_weights)
            loss.backward()
            optimizer.step()

    model.eval()
    all_pred, all_true = [], []
    with torch.no_grad():
        for batch in test_loader:
            pred = model(batch.x, batch.edge_index, batch.batch).argmax(dim=1)
            all_pred.append(pred)
            all_true.append(batch.y)
    pred = torch.cat(all_pred)
    true = torch.cat(all_true)

    acc = (pred == true).float().mean().item()
    tp = ((pred == 1) & (true == 1)).sum().item()
    fp = ((pred == 1) & (true == 0)).sum().item()
    fn = ((pred == 0) & (true == 1)).sum().item()
    tn = ((pred == 0) & (true == 0)).sum().item()
    precision = tp / (tp + fp) if (tp + fp) else 0.0
    recall = tp / (tp + fn) if (tp + fn) else 0.0
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) else 0.0
    test_class_counts = torch.bincount(true, minlength=2).tolist()
    return model, {'accuracy': acc, 'precision': precision, 'recall': recall, 'f1': f1,
            'confusion_matrix': {'tp': tp, 'fp': fp, 'fn': fn, 'tn': tn},
            'test_class_counts': {'normal': test_class_counts[0], 'anomalous': test_class_counts[1]}}


skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
fold_results = []
batadal_bench_model, batadal_bench_test_graphs = None, None  # kept from last fold, for Part H size/latency benchmarking
for fold, (train_idx, test_idx) in enumerate(skf.split(window_graphs, window_labels)):
    train_graphs = [window_graphs[i] for i in train_idx]
    test_graphs = [window_graphs[i] for i in test_idx]
    fold_model, metrics = run_fold(train_graphs, test_graphs)
    fold_results.append(metrics)
    batadal_bench_model, batadal_bench_test_graphs = fold_model, test_graphs
    print(f"Fold {fold + 1}: acc={metrics['accuracy']:.4f} "
          f"prec={metrics['precision']:.4f} rec={metrics['recall']:.4f} "
          f"f1={metrics['f1']:.4f}  (test windows={len(test_idx)}, "
          f"positives={int(window_labels[test_idx].sum())})")
````

````python
# Aggregate cross-validated BATADAL results
import numpy as np

accs = np.array([r['accuracy'] for r in fold_results])
precs = np.array([r['precision'] for r in fold_results])
recs = np.array([r['recall'] for r in fold_results])
f1s = np.array([r['f1'] for r in fold_results])

batadal_cv_summary = {
    'accuracy_mean': accs.mean(), 'accuracy_std': accs.std(),
    'precision_mean': precs.mean(), 'precision_std': precs.std(),
    'recall_mean': recs.mean(), 'recall_std': recs.std(),
    'f1_mean': f1s.mean(), 'f1_std': f1s.std(),
    'n_windows': len(window_graphs), 'n_positive_windows': int(window_labels.sum()),
    'n_folds': len(fold_results),
}

print("=" * 60)
print("BATADAL CROSS-DATASET VALIDATION -- 5-fold stratified CV")
print("=" * 60)
print(f"Windows          : {batadal_cv_summary['n_windows']} "
      f"({batadal_cv_summary['n_positive_windows']} positive)")
print(f"Accuracy         : {accs.mean():.4f} +/- {accs.std():.4f}")
print(f"Precision        : {precs.mean():.4f} +/- {precs.std():.4f}")
print(f"Recall           : {recs.mean():.4f} +/- {recs.std():.4f}")
print(f"F1               : {f1s.mean():.4f} +/- {f1s.std():.4f}")

print("\n" + "=" * 60)
print("SUMMARY -- ICS-Flow vs BATADAL")
print("=" * 60)
print(f"{'Metric':<12}{'ICS-Flow (edge-level)':<26}{'BATADAL (window-level, 5-fold)':<32}")
print(f"{'Accuracy':<12}{edge_metrics['test_accuracy']:<26.4f}{accs.mean():.4f} +/- {accs.std():.4f}")
print(f"{'Precision':<12}{edge_metrics['precision']:<26.4f}{precs.mean():.4f} +/- {precs.std():.4f}")
print(f"{'Recall':<12}{edge_metrics['recall']:<26.4f}{recs.mean():.4f} +/- {recs.std():.4f}")
print(f"{'F1':<12}{edge_metrics['f1']:<26.4f}{f1s.mean():.4f} +/- {f1s.std():.4f}")
````

````python
# Whitelist of labels/edge types actually produced by build_knowledge_graph() -- used to validate before interpolating into Cypher, since node_type/edge_type
# drive dynamic label/relationship names below.
VALID_NODE_LABELS = {"Asset", "Attacker", "Technique", "CVE", "Alert"}
VALID_EDGE_TYPES = {"COMMUNICATES_WITH", "USES_TECHNIQUE", "TARGETS",
                     "VULNERABLE_TO", "ENABLES", "TRIGGERED_ON"}

nodes_df = pd.read_csv("week4_kg_nodes.csv")
edges_df = pd.read_csv("week4_kg_edges.csv")

bad_nodes = set(nodes_df["node_type"].dropna().unique()) - VALID_NODE_LABELS
bad_edges = set(edges_df["edge_type"].dropna().unique()) - VALID_EDGE_TYPES
if bad_nodes or bad_edges:
    raise ValueError(f"Unexpected node_type/edge_type values found: "
                      f"nodes={bad_nodes}, edges={bad_edges}. "
                      f"Update the whitelist above before loading into Neo4j.")

driver = GraphDatabase.driver(NEO4J_URI, auth=(NEO4J_USER, NEO4J_PASSWORD))

def load_nodes(tx, nodes_df):
    for _, row in nodes_df.iterrows():
        label = row["node_type"] if pd.notna(row["node_type"]) and row["node_type"] else "Unknown"
        tx.run(
            f"MERGE (n:{label} {{id: $id}}) "
            "SET n.label = $label_name, n.role = $role, n.risk_score = $risk_score",
            id=row["id"], label_name=row["label"], role=row.get("role", ""),
            risk_score=float(row["risk_score"]) if pd.notna(row["risk_score"]) else 0.0,
        )

def load_edges(tx, edges_df):
    for _, row in edges_df.iterrows():
        etype = row["edge_type"] if pd.notna(row["edge_type"]) and row["edge_type"] else "RELATED_TO"
        tx.run(
            f"MATCH (a {{id: $source}}), (b {{id: $target}}) "
            f"MERGE (a)-[r:{etype}]->(b) "
            "SET r.attack_type = $attack_type, r.count = $count, "
            "r.weight = $weight, r.cvss = $cvss",
            source=row["source"], target=row["target"],
            attack_type=row.get("attack_type", "") if pd.notna(row.get("attack_type")) else "",
            count=float(row["count"]) if pd.notna(row.get("count")) else 0,
            weight=float(row["weight"]) if pd.notna(row.get("weight")) else 0,
            cvss=float(row["cvss"]) if pd.notna(row.get("cvss")) else 0,
        )

with driver.session() as session:
    session.execute_write(load_nodes, nodes_df)
    session.execute_write(load_edges, edges_df)

print(f"Loaded {len(nodes_df)} nodes and {len(edges_df)} edges into Neo4j.")
driver.close()
````

### G.1 — Baseline model (topology only: `Source`, `Destination`, `Attack_Type`)

IP-pair topology and a constant, non-identifying node feature vector only -- no per-flow packet features, no raw flow-count degree.


````python
# Baseline topology: IP-pair graph + flow-level query edges
import torch
import numpy as np
from torch_geometric.data import Data

all_ips = sorted(set(scada_df[SRC_IP_COL].astype(str)) | set(scada_df[DST_IP_COL].astype(str)))
scada_ip_index = {ip: i for i, ip in enumerate(all_ips)}

# Node features: constant, non-identifying placeholder vector. Raw flow-count
# degree is deliberately excluded here -- spoofed flood source IPs and
# genuine hosts occupy disjoint, trivially separable regions of degree-space
# in this dataset, so a degree-based feature would let the classifier solve
# detection via traffic volume alone rather than per-flow behavior. A
# constant feature forces the GraphSAGE embedder to rely only on graph
# structure (message passing over the topology below), not on how many
# times an IP appears.
scada_x = torch.ones((len(all_ips), 4), dtype=torch.float32)

# Fixed topology: deduplicated IP-pair communication graph
unique_pairs = scada_df[[SRC_IP_COL, DST_IP_COL]].astype(str).drop_duplicates()
topo_src = unique_pairs[SRC_IP_COL].map(scada_ip_index).to_numpy()
topo_dst = unique_pairs[DST_IP_COL].map(scada_ip_index).to_numpy()
scada_edge_index = torch.tensor(np.stack([topo_src, topo_dst]), dtype=torch.long)

# Query edges: every individual flow record (with repeats) -- the classification targets
q_src = torch.tensor(scada_df[SRC_IP_COL].astype(str).map(scada_ip_index).to_numpy(), dtype=torch.long)
q_dst = torch.tensor(scada_df[DST_IP_COL].astype(str).map(scada_ip_index).to_numpy(), dtype=torch.long)
scada_query_edges = torch.stack([q_src, q_dst])
scada_y = torch.tensor(scada_df["is_attack"].to_numpy(), dtype=torch.long)

scada_data = Data(x=scada_x, edge_index=scada_edge_index)
print(f"SCADANet: {len(all_ips)} IPs, {scada_edge_index.shape[1]} unique topology edges, "
      f"{scada_query_edges.shape[1]} flow records, "
      f"{int(scada_y.sum())} attack flows ({scada_y.float().mean()*100:.1f}%)")

# Sanity check: IP-level class separability -- confirms why an IP-held-out
# split (rather than a random flow split) is required for a meaningful result.
ip_attack_rate = pd.concat([
    scada_df.groupby(SRC_IP_COL)["is_attack"].mean(),
    scada_df.groupby(DST_IP_COL)["is_attack"].mean()
]).groupby(level=0).mean()
n_pure = int(((ip_attack_rate == 0) | (ip_attack_rate == 1)).sum())
print(f"IPs with attack rate exactly 0 or 1: {n_pure} / {len(ip_attack_rate)}")

# Trivial degree-only rule -- quantifies how much signal raw flow-count alone
# would provide, without using it as an actual model feature. Used only as
# a reference point for interpreting the GNN's results below.
out_deg = scada_df.groupby(SRC_IP_COL).size()
deg_by_node = torch.tensor([out_deg.get(ip, 0) for ip in all_ips], dtype=torch.float32)
trivial_pred = (deg_by_node[q_src] < 5).long()
tp = ((trivial_pred == 1) & (scada_y == 1)).sum().item()
fp = ((trivial_pred == 1) & (scada_y == 0)).sum().item()
fn = ((trivial_pred == 0) & (scada_y == 1)).sum().item()
trivial_precision = tp / (tp + fp) if (tp + fp) else 0.0
trivial_recall = tp / (tp + fn) if (tp + fn) else 0.0
print(f"Reference -- trivial degree-only rule (out_degree < 5): "
      f"precision={trivial_precision:.4f}, recall={trivial_recall:.4f}")
````

````python
# Train + evaluate the baseline GraphSAGE EdgeClassifier (IP-held-out split)
def train_eval_scadanet_ip_split(data, query_edges, y, all_ips, epochs=200, test_ratio=0.2, seed=42):
    n_ips = len(all_ips)
    g = torch.Generator().manual_seed(seed)
    perm = torch.randperm(n_ips, generator=g)
    n_test_ips = max(1, int(n_ips * test_ratio))
    test_ip_idx = set(perm[:n_test_ips].tolist())

    src, dst = query_edges[0], query_edges[1]
    is_test_flow = torch.tensor([
        (s.item() in test_ip_idx) or (d.item() in test_ip_idx)
        for s, d in zip(src, dst)
    ])
    test_idx = is_test_flow.nonzero(as_tuple=True)[0]
    train_idx = (~is_test_flow).nonzero(as_tuple=True)[0]

    train_y = y[train_idx]
    class_counts = torch.bincount(train_y, minlength=2).float()
    class_weights = train_y.shape[0] / (2.0 * class_counts.clamp(min=1))
    print(f"Class counts (train): normal={int(class_counts[0])}, "
          f"attack={int(class_counts[1])}  ->  weights={class_weights.tolist()}")
    print(f"Held-out IPs: {n_test_ips} / {n_ips} "
          f"({len(train_idx)} train flows, {len(test_idx)} test flows)")

    model = EdgeClassifier(in_dim=data.x.shape[1])
    optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=5e-4)

    model.train()
    for epoch in range(epochs):
        optimizer.zero_grad()
        out = model(data.x, data.edge_index, query_edges[:, train_idx])
        loss = F.cross_entropy(out, train_y, weight=class_weights)
        loss.backward()
        optimizer.step()

    model.eval()
    with torch.no_grad():
        test_pred = model(data.x, data.edge_index, query_edges[:, test_idx]).argmax(dim=1)
        test_y = y[test_idx]
        acc = (test_pred == test_y).float().mean().item()
        tp = ((test_pred == 1) & (test_y == 1)).sum().item()
        fp = ((test_pred == 1) & (test_y == 0)).sum().item()
        fn = ((test_pred == 0) & (test_y == 1)).sum().item()
        tn = ((test_pred == 0) & (test_y == 0)).sum().item()
        precision = tp / (tp + fp) if (tp + fp) else 0.0
        recall = tp / (tp + fn) if (tp + fn) else 0.0
        f1 = 2 * precision * recall / (precision + recall) if (precision + recall) else 0.0

    return model, {
        "test_accuracy": acc, "precision": precision, "recall": recall, "f1": f1,
        "n_train_edges": len(train_idx), "n_test_edges": len(test_idx),
        "n_test_ips": n_test_ips, "n_total_ips": n_ips,
        "n_features_used": data.x.shape[1],
        "confusion_matrix": {"tp": tp, "fp": fp, "fn": fn, "tn": tn},
        "test_class_counts": {"normal": int((test_y == 0).sum()), "attack": int((test_y == 1).sum())},
    }

scadanet_model, scadanet_metrics_baseline = train_eval_scadanet_ip_split(
    scada_data, scada_query_edges, scada_y, all_ips
)
print("\nSCADANet baseline (3 columns, IP-held-out split):", scadanet_metrics_baseline)
````

````python
# Step 2 — classify columns based on the inspection above

# Fully NaN or constant -> zero information, drop. Detected directly from
# the inspection (nunique <= 1) rather than hardcoded, so this stays correct
# if the underlying file changes.
DEAD_COLS = [c for c in scada_df.columns if nunique[c] <= 1]

# Direct/near-direct encodings of the label -> must never be features
LEAK_COLS = ["Traffic_Category", "Attack_Type"]

# Identifiers, free text, and near-unique fingerprints -> excluded from the
# feature vector. Source/Destination already drive topology; the JA4*
# fingerprints are near-unique per flow and risk identity memorization
# rather than generalizable signal
ID_AND_FREE_TEXT_COLS = [
    "Source", "Destination", "Info", "Time", "Tcp_payload",
    "Tcp_ack", "Tcp_ack.1", "Tcp_seq", "Tcp_src_port",
    "Udp_srcport", "Udp_dstport", "Tcp_stream",
    "JA4L", "JA4LS", "JA4S", "JA4S_R", "JA4T", "JA4TS",
    "JA4H", "JA4H_R", "JA4H_RO", "JA4X", "JA4X_R",
    "Length",
]

# High-cardinality free-text columns, held out deliberately: raw one-hot on
# these would add ~185 mostly-sparse dimensions on top of ~535K flows split
# across small attack subclasses. The model would likely memorize
# "this exact URI/payload = attack" rather than learn a generalizable
# pattern. Usable later with structural extraction (URI length, path depth,
# presence of query string, payload byte-length) rather than raw identity.
EXCLUDED_HIGH_CARDINALITY_COLS = ["Http_request_uri", "Http_file_data"]

# Real per-flow signal -> the feature set
NUMERIC_FEATURE_COLS = [
    "ip_ttl", "frame_len", "ip_len", "Tcp_window_size", "Udp_length",
    "Tcp_len", "Tcp_hdr_len", "Http_content_length", "Http_time",
    "Frame_time_delta",
]
CATEGORICAL_FEATURE_COLS = [
    "Protocol", "Http_request_method", "Http_response_code",
    "Tcp_flags_reset", "http_accept", "Icmp_type", "Icmp_code",
    "Http_user_agent",         # 12 unique -- scanner/tool signatures (e.g. Nmap)
    "Tls_record",               # 2 unique -- presence flag
    "tls_handshake",            # 2 unique -- presence flag
    "Http_request_uri_query",   # 2 unique -- presence flag
    "Time_response",            # 2 unique -- presence flag
    "Tcp_analysis_flags",       # 2 unique -- presence flag
]

n_raw_used = len(NUMERIC_FEATURE_COLS) + len(CATEGORICAL_FEATURE_COLS)
print("--- Column accounting (60 total raw columns) ---")
print(f"Dead/constant, dropped:                  {len(DEAD_COLS)}")
print(f"Label / leak columns, dropped:            {len(LEAK_COLS)}")
print(f"IDs / free text / fingerprints, dropped:  {len(ID_AND_FREE_TEXT_COLS)}")
print(f"High-cardinality, deliberately held out:  {len(EXCLUDED_HIGH_CARDINALITY_COLS)}")
print(f"Raw columns used as features:             {n_raw_used} "
      f"({len(NUMERIC_FEATURE_COLS)} numeric + {len(CATEGORICAL_FEATURE_COLS)} categorical)")
````

````python
# Step 3 — build the per-flow edge feature matrix from the surviving columns
def normalize_mixed_type_categorical(series):
    """Collapse numeric-like values written in different forms (e.g. the
    float 200.0, the string '200', and the string '200.0' in
    Http_response_code) into one consistent string before one-hot encoding.
    Without this, pd.get_dummies treats them as three separate categories
    and dilutes what should be a single, clean signal."""
    numeric = pd.to_numeric(series, errors="coerce")
    is_numeric_like = numeric.notna()
    normalized = series.astype(str)
    normalized[is_numeric_like] = numeric[is_numeric_like].astype(int).astype(str)
    normalized[series.isna()] = "missing"
    return normalized


def build_edge_attr(df, numeric_cols, categorical_cols):
    numeric_cols = [c for c in numeric_cols if c in df.columns]
    categorical_cols = [c for c in categorical_cols if c in df.columns]

    # Numeric: coerce, median-impute, z-score normalize
    num_block = df[numeric_cols].apply(pd.to_numeric, errors="coerce")
    num_block = num_block.fillna(num_block.median(numeric_only=True))
    num_arr = num_block.to_numpy(dtype=np.float32)
    mu, sigma = num_arr.mean(axis=0, keepdims=True), num_arr.std(axis=0, keepdims=True)
    sigma[sigma == 0] = 1.0
    num_arr = (num_arr - mu) / sigma

    # Categorical: normalize mixed-type values, fill missing, one-hot
    cat_block = df[categorical_cols].apply(normalize_mixed_type_categorical)
    cat_dummies = pd.get_dummies(cat_block, columns=categorical_cols, dummy_na=False)
    cat_arr = cat_dummies.to_numpy(dtype=np.float32)

    edge_attr = np.concatenate([num_arr, cat_arr], axis=1)
    feature_names = numeric_cols + list(cat_dummies.columns)
    return torch.tensor(edge_attr, dtype=torch.float32), feature_names

scada_edge_attr, edge_feature_names = build_edge_attr(
    scada_df, NUMERIC_FEATURE_COLS, CATEGORICAL_FEATURE_COLS
)
print(f"Edge feature matrix: {scada_edge_attr.shape} "
      f"({len(edge_feature_names)} feature dimensions after one-hot expansion)")
````

````python
# Step 4 — GraphSAGE EdgeClassifier extended to consume per-flow edge features.

class EdgeClassifierWithAttr(torch.nn.Module):
    def __init__(self, in_dim, edge_attr_dim, hidden_dim=16, edge_proj_dim=16):
        super().__init__()
        self.embedder = SAGEEmbedder(in_dim, hidden_dim, out_dim=hidden_dim)
        self.edge_proj = torch.nn.Sequential(
            torch.nn.Linear(edge_attr_dim, edge_proj_dim),
            torch.nn.ReLU(),
        )
        self.head = torch.nn.Sequential(
            torch.nn.Linear(hidden_dim * 2 + edge_proj_dim, hidden_dim),
            torch.nn.ReLU(),
            torch.nn.Dropout(p=0.3),
            torch.nn.Linear(hidden_dim, 2),
        )

    def forward(self, x, edge_index, query_edges, query_edge_attr):
        h = self.embedder(x, edge_index)
        edge_repr = self.edge_proj(query_edge_attr)
        combined = torch.cat(
            [h[query_edges[0]], h[query_edges[1]], edge_repr], dim=1
        )
        return self.head(combined)


def train_eval_scadanet_enriched(data, query_edges, edge_attr, y, all_ips,
                                  epochs=400, test_ratio=0.2, seed=42,
                                  log_every=20, lr=0.003, grad_clip=1.0):
    n_ips = len(all_ips)
    g = torch.Generator().manual_seed(seed)
    perm = torch.randperm(n_ips, generator=g)
    n_test_ips = max(1, int(n_ips * test_ratio))
    test_ip_idx = set(perm[:n_test_ips].tolist())

    src, dst = query_edges[0], query_edges[1]
    is_test_flow = torch.tensor([
        (s.item() in test_ip_idx) or (d.item() in test_ip_idx)
        for s, d in zip(src, dst)
    ])
    test_idx = is_test_flow.nonzero(as_tuple=True)[0]
    train_idx = (~is_test_flow).nonzero(as_tuple=True)[0]

    train_y = y[train_idx]
    test_y = y[test_idx]
    class_counts = torch.bincount(train_y, minlength=2).float()
    class_weights = train_y.shape[0] / (2.0 * class_counts.clamp(min=1))
    print(f"Class counts (train): normal={int(class_counts[0])}, "
          f"attack={int(class_counts[1])}  ->  weights={class_weights.tolist()}")
    print(f"Held-out IPs: {n_test_ips} / {n_ips} "
          f"({len(train_idx)} train flows, {len(test_idx)} test flows)")

    model = EdgeClassifierWithAttr(in_dim=data.x.shape[1], edge_attr_dim=edge_attr.shape[1])
    optimizer = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=5e-4)

    def compute_metrics(pred, true_y):
        tp = ((pred == 1) & (true_y == 1)).sum().item()
        fp = ((pred == 1) & (true_y == 0)).sum().item()
        fn = ((pred == 0) & (true_y == 1)).sum().item()
        tn = ((pred == 0) & (true_y == 0)).sum().item()
        acc = (pred == true_y).float().mean().item()
        precision = tp / (tp + fp) if (tp + fp) else 0.0
        recall = tp / (tp + fn) if (tp + fn) else 0.0
        f1 = 2 * precision * recall / (precision + recall) if (precision + recall) else 0.0
        return acc, precision, recall, f1, {"tp": tp, "fp": fp, "fn": fn, "tn": tn}

    history = []

    model.train()
    loss_history = []
    for epoch in range(epochs):
        optimizer.zero_grad()
        out = model(data.x, data.edge_index, query_edges[:, train_idx], edge_attr[train_idx])
        loss = F.cross_entropy(out, train_y, weight=class_weights)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=grad_clip)
        optimizer.step()
        loss_history.append(loss.item())

        if (epoch + 1) % log_every == 0 or epoch == 0:
            model.eval()
            with torch.no_grad():
                train_pred = out.argmax(dim=1)
                tr_acc, tr_p, tr_r, tr_f1, _ = compute_metrics(train_pred, train_y)

                test_out = model(data.x, data.edge_index, query_edges[:, test_idx], edge_attr[test_idx])
                test_pred = test_out.argmax(dim=1)
                te_acc, te_p, te_r, te_f1, _ = compute_metrics(test_pred, test_y)
            model.train()

            history.append({"epoch": epoch + 1, "loss": loss.item(),
                             "train": {"acc": tr_acc, "precision": tr_p, "recall": tr_r, "f1": tr_f1},
                             "test": {"acc": te_acc, "precision": te_p, "recall": te_r, "f1": te_f1}})

            print(f"epoch {epoch + 1:4d}/{epochs}  loss={loss.item():.4f}  |  "
                  f"TRAIN acc={tr_acc:.4f} prec={tr_p:.4f} rec={tr_r:.4f} f1={tr_f1:.4f}  |  "
                  f"TEST acc={te_acc:.4f} prec={te_p:.4f} rec={te_r:.4f} f1={te_f1:.4f}")

    model.eval()
    with torch.no_grad():
        test_pred = model(data.x, data.edge_index, query_edges[:, test_idx], edge_attr[test_idx]).argmax(dim=1)
        acc, precision, recall, f1, cm = compute_metrics(test_pred, test_y)

    return model, {
        "test_accuracy": acc, "precision": precision, "recall": recall, "f1": f1,
        "n_train_edges": len(train_idx), "n_test_edges": len(test_idx),
        "n_test_ips": n_test_ips, "n_total_ips": n_ips,
        "n_raw_columns_used": n_raw_used,
        "n_features_used": edge_attr.shape[1],
        "final_train_loss": loss_history[-1],
        "confusion_matrix": cm,
        "test_class_counts": {"normal": int((test_y == 0).sum()), "attack": int((test_y == 1).sum())},
        "epoch_history": history,
        "test_idx": test_idx,
    }

scadanet_model_enriched, scadanet_metrics_enriched = train_eval_scadanet_enriched(
    scada_data, scada_query_edges, scada_edge_attr, scada_y, all_ips
)
print("\nSCADANet enriched (23 raw columns, IP-held-out split):", scadanet_metrics_enriched)
````

````python
def get_ip_diversity_per_attack_type(df, label_col, src_col):
    return df.groupby(label_col)[src_col].nunique().sort_values()

ip_diversity = get_ip_diversity_per_attack_type(scada_df, LABEL_COL, SRC_IP_COL)
print("Distinct source IPs per attack type:")
print(ip_diversity)

# Attack types with many distinct source IPs can be fairly evaluated under
# an IP-held-out split (Track A). Anything below the threshold is generated
# by a small, fixed set of attacker IPs and cannot be split this way.
MULTI_IP_ATTACK_TYPES = ip_diversity[ip_diversity > 50].index.tolist()
MULTI_IP_ATTACK_TYPES = [t for t in MULTI_IP_ATTACK_TYPES if "normal" not in t.strip().lower()]

SINGLE_IP_ATTACK_TYPES = [t for t in scada_df[LABEL_COL].unique()
                          if t not in MULTI_IP_ATTACK_TYPES
                          and "normal" not in t.strip().lower()]

print(f"\nTrack A (IP-held-out) covers : {MULTI_IP_ATTACK_TYPES}")
print(f"Track B (flow-level) covers  : {SINGLE_IP_ATTACK_TYPES}")
````

````python
def build_flow_level_split_idx(df, label_col, attack_types, test_ratio=0.2, seed=42):
    """Random flow-level split restricted to single-fixed-IP attack types
    plus all normal traffic. NOT IP-held-out -- the same attacker IP can
    (and will) appear in both train and test here. That's intentional:
    this track answers 'can the model recognize more of a known attacker's
    behavior', not 'does it generalize to unseen attackers'."""
    is_target_attack = df[label_col].isin(attack_types)
    is_normal = df[label_col].str.strip().str.lower().str.contains("normal")
    subset_mask = (is_target_attack | is_normal).to_numpy()
    subset_idx = np.nonzero(subset_mask)[0]

    g = torch.Generator().manual_seed(seed)
    perm = torch.randperm(len(subset_idx), generator=g).numpy()
    n_test = max(1, int(len(subset_idx) * test_ratio))
    test_pos = subset_idx[perm[:n_test]]
    train_pos = subset_idx[perm[n_test:]]
    return torch.tensor(train_pos, dtype=torch.long), torch.tensor(test_pos, dtype=torch.long)

train_idx_b, test_idx_b = build_flow_level_split_idx(scada_df, LABEL_COL, SINGLE_IP_ATTACK_TYPES)
print(f"Track B: {len(train_idx_b)} train flows, {len(test_idx_b)} test flows "
      f"across {len(SINGLE_IP_ATTACK_TYPES)} single-IP attack types + normal traffic")
````

````python
def compute_subtype_balanced_weights(df, label_col, idx):
    """Weight every individual label (each attack subtype + normal) inversely
    to its own frequency in the training set, so rare attack subtypes
    (scans, protocol abuse) get comparable gradient signal to dominant ones
    (floods), instead of being drowned out."""
    labels_subset = df.iloc[idx.numpy()][label_col]
    counts = labels_subset.value_counts()
    n = len(labels_subset)
    n_classes = len(counts)
    weight_per_label = {label: n / (n_classes * count) for label, count in counts.items()}
    return weight_per_label


def train_eval_scadanet_flowlevel(data, query_edges, edge_attr, y, train_idx, test_idx,
                                      df, label_col, epochs=400, log_every=20,
                                      lr=0.003, grad_clip=1.0, seed=42):
    torch.manual_seed(seed)

    train_y = y[train_idx]
    test_y = y[test_idx]

    # Per-flow subtype-balanced weights, computed once on the train split
    subtype_weights = compute_subtype_balanced_weights(df, label_col, train_idx)
    print("Per-subtype training weights:")
    for label, w in sorted(subtype_weights.items(), key=lambda x: -x[1]):
        print(f"  {label:<20} weight={w:.4f}")

    train_labels = df.iloc[train_idx.numpy()][label_col].values
    per_flow_weight = torch.tensor(
        [subtype_weights[l] for l in train_labels], dtype=torch.float32
    )

    model = EdgeClassifierWithAttr(in_dim=data.x.shape[1], edge_attr_dim=edge_attr.shape[1])
    optimizer = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=5e-4)

    def compute_metrics(pred, true_y):
        tp = ((pred == 1) & (true_y == 1)).sum().item()
        fp = ((pred == 1) & (true_y == 0)).sum().item()
        fn = ((pred == 0) & (true_y == 1)).sum().item()
        tn = ((pred == 0) & (true_y == 0)).sum().item()
        acc = (pred == true_y).float().mean().item()
        precision = tp / (tp + fp) if (tp + fp) else 0.0
        recall = tp / (tp + fn) if (tp + fn) else 0.0
        f1 = 2 * precision * recall / (precision + recall) if (precision + recall) else 0.0
        return acc, precision, recall, f1, {"tp": tp, "fp": fp, "fn": fn, "tn": tn}

    model.train()
    loss_history = []
    for epoch in range(epochs):
        optimizer.zero_grad()
        out = model(data.x, data.edge_index, query_edges[:, train_idx], edge_attr[train_idx])

        # Per-sample subtype-balanced loss, instead of a single binary class weight
        per_sample_loss = F.cross_entropy(out, train_y, reduction='none')
        loss = (per_sample_loss * per_flow_weight).mean()

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=grad_clip)
        optimizer.step()
        loss_history.append(loss.item())

        if (epoch + 1) % log_every == 0 or epoch == 0:
            model.eval()
            with torch.no_grad():
                train_pred = out.argmax(dim=1)
                tr_acc, tr_p, tr_r, tr_f1, _ = compute_metrics(train_pred, train_y)
                test_out = model(data.x, data.edge_index, query_edges[:, test_idx], edge_attr[test_idx])
                test_pred = test_out.argmax(dim=1)
                te_acc, te_p, te_r, te_f1, _ = compute_metrics(test_pred, test_y)
            model.train()
            print(f"epoch {epoch + 1:4d}/{epochs}  loss={loss.item():.4f}  |  "
                  f"TRAIN acc={tr_acc:.4f} prec={tr_p:.4f} rec={tr_r:.4f} f1={tr_f1:.4f}  |  "
                  f"TEST acc={te_acc:.4f} prec={te_p:.4f} rec={te_r:.4f} f1={te_f1:.4f}")

    model.eval()
    with torch.no_grad():
        test_pred = model(data.x, data.edge_index, query_edges[:, test_idx], edge_attr[test_idx]).argmax(dim=1)
        acc, precision, recall, f1, cm = compute_metrics(test_pred, test_y)

    return model, {
        "test_accuracy": acc, "precision": precision, "recall": recall, "f1": f1,
        "n_train_edges": len(train_idx), "n_test_edges": len(test_idx),
        "confusion_matrix": cm,
        "test_class_counts": {"normal": int((test_y == 0).sum()), "attack": int((test_y == 1).sum())},
        "test_idx": test_idx,
        "subtype_weights": subtype_weights,
    }


scadanet_model_trackb, scadanet_metrics_trackb = train_eval_scadanet_flowlevel(
    scada_data, scada_query_edges, scada_edge_attr, scada_y,
    train_idx_b, test_idx_b, scada_df, LABEL_COL
)
print("\nSCADANet Track B  (subtype-balanced):", scadanet_metrics_trackb)

# Per-attack-type breakdown
with torch.no_grad():
    final_pred_b2 = scadanet_model_trackb(
        scada_data.x, scada_data.edge_index,
        scada_query_edges[:, test_idx_b], scada_edge_attr[test_idx_b]
    ).argmax(dim=1)

test_types_b2 = scada_df.iloc[test_idx_b.numpy()][LABEL_COL].values
print("\n" + "=" * 50)
print("TRACK B  (SUBTYPE-BALANCED) -- PER-ATTACK-TYPE BREAKDOWN")
print("=" * 50)
for atype in pd.unique(test_types_b2):
    mask = test_types_b2 == atype
    y_t = scada_y[test_idx_b][mask]
    p_t = final_pred_b2[mask]
    n = mask.sum()
    if "normal" in atype.strip().lower():
        fpr = ((p_t == 1) & (y_t == 0)).sum().item() / max(n, 1)
        print(f"{atype:<20} n={n:<8} FPR={fpr:.4f}")
    else:
        tp = ((p_t == 1) & (y_t == 1)).sum().item()
        fn = ((p_t == 0) & (y_t == 1)).sum().item()
        recall = tp / (tp + fn) if (tp + fn) else float('nan')
        print(f"{atype:<20} n={n:<8} recall={recall:.4f}")
````

### **G.4 — Interpretation: Why Track A and Track B Disagree**

The Week 6 vs. Week 7 aggregate comparison (recall = 1.0000 for both baseline
and enriched models) is misleading if reported on its own. Investigating the
underlying graph structure revealed that 10 of the 14 attack types in
SCADANet are generated by a single, fixed attacker IP each, while the
remaining types (`tcp_syn_flood`, `udp_flood`) use spoofed source IPs that
each appear only 1–2 times in the whole dataset. Because the baseline node
features are raw flow-count degree (`[out_flows, in_flows]`), spoofed flood
sources are trivially separable from normal hosts by degree alone —
independent of any real traffic-content signal.

This means the standard IP-held-out split (**Track A**) only ever evaluates
on `tcp_syn_flood` / `udp_flood` (the two multi-IP attack types), and its
recall = 1.0000 largely reflects this degree artifact rather than
generalization to new attacker behavior. With only 14 `udp_flood` test
flows, Track A is also too small a sample to draw strong conclusions from
on its own.

To separate this shortcut from genuine model performance, evaluation was
split into two tracks:

- **Track A (IP-held-out)** — tests generalization to unseen attacker
  infrastructure. Only valid for `tcp_syn_flood` / `udp_flood`, where
  enough distinct attacker IPs exist to hold some out.
- **Track B (flow-level, subtype-balanced)** — tests recognition of known
  attacker *behavior* for the 12 single-fixed-IP attack types
  (`sql_injection`, `insider_threat`, `modbus_abuse`, `apt_exfil`, etc.),
  where an IP-held-out split is structurally impossible. The same attacker
  IP may appear in both train and test here by design — this track answers
  a different, narrower question than Track A.

Track B (105,711 test flows, 88,839 of them attacks) is the more
informative result of the two: recall = 0.9997 across a much larger and
more diverse sample, achieved using the 224-dimension content-feature set
rather than IP-degree alone. It is not confounded by the spoofing artifact
in the same way Track A is.

**Limitation identified but not yet resolved:** Track B's precision
(0.9508) is noticeably lower than Track A's (0.9853), driven by a 27.2%
false-positive rate on normal traffic — over a quarter of held-out normal
flows are misclassified as attacks. This is the main open weakness of the
enriched model and the priority for further investigation (e.g. auditing
which normal flows are misclassified, and whether specific features or
hosts are driving it), rather than the near-saturated recall figures,
which no longer discriminate between genuine and shortcut-driven
performance at this point.

````python
# Parameter count + model size helper
import torch
import os
import time

def model_stats(model, name):
    n_params = sum(p.numel() for p in model.parameters())
    n_trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    tmp_path = f"_tmp_{name}.pt"
    torch.save(model.state_dict(), tmp_path)
    size_mb = os.path.getsize(tmp_path) / (1024 ** 2)
    os.remove(tmp_path)
    return {"n_params": n_params, "n_trainable_params": n_trainable, "size_mb": round(size_mb, 4)}

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Benchmarking device: {device}")
````

````python
# Inference latency helpers (mean over repeated runs)
def bench_edge_model_latency(model, data, query_edges, n_runs=50, warmup=5):
    """Mean/std latency (ms) for one full forward pass over query_edges."""
    model.eval()
    with torch.no_grad():
        for _ in range(warmup):
            model(data.x, data.edge_index, query_edges)
        times = []
        for _ in range(n_runs):
            t0 = time.perf_counter()
            model(data.x, data.edge_index, query_edges)
            times.append((time.perf_counter() - t0) * 1000)
    times = np.array(times)
    return {"mean_ms": float(times.mean()), "std_ms": float(times.std()),
            "n_edges": query_edges.shape[1],
            "ms_per_edge": float(times.mean() / query_edges.shape[1])}


def bench_edge_model_with_attr_latency(model, data, query_edges, edge_attr, n_runs=50, warmup=5):
    """Same as above, for the edge-feature-aware EdgeClassifierWithAttr."""
    model.eval()
    with torch.no_grad():
        for _ in range(warmup):
            model(data.x, data.edge_index, query_edges, edge_attr)
        times = []
        for _ in range(n_runs):
            t0 = time.perf_counter()
            model(data.x, data.edge_index, query_edges, edge_attr)
            times.append((time.perf_counter() - t0) * 1000)
    times = np.array(times)
    return {"mean_ms": float(times.mean()), "std_ms": float(times.std()),
            "n_edges": query_edges.shape[1],
            "ms_per_edge": float(times.mean() / query_edges.shape[1])}


def bench_graph_model_latency(model, test_graphs, n_runs=50, warmup=5):
    """Mean/std latency (ms) for one full forward pass over a batch of graphs."""
    from torch_geometric.loader import DataLoader
    loader = DataLoader(test_graphs, batch_size=len(test_graphs), shuffle=False)
    batch = next(iter(loader))
    model.eval()
    with torch.no_grad():
        for _ in range(warmup):
            model(batch.x, batch.edge_index, batch.batch)
        times = []
        for _ in range(n_runs):
            t0 = time.perf_counter()
            model(batch.x, batch.edge_index, batch.batch)
            times.append((time.perf_counter() - t0) * 1000)
    times = np.array(times)
    return {"mean_ms": float(times.mean()), "std_ms": float(times.std()),
            "n_graphs": len(test_graphs),
            "ms_per_graph": float(times.mean() / len(test_graphs))}
````

#### **Validation-Based Threshold Calibration**

The original **Track B model (`scadanet_model_trackb`)** is kept fixed throughout this experiment, and **no retraining** is performed.

A validation subset sampled from the training data (`train_idx_b`) is used **only** to select the optimal decision threshold. The selected threshold is then applied **once** to the original model's existing test-set probabilities (`test_probs`).

This ensures that the comparison isolates the effect of **threshold calibration**, while keeping the trained model and test predictions unchanged. No test-set labels are used during threshold selection, preventing test-set leakage.

````python
from sklearn.metrics import precision_recall_curve
import torch

g = torch.Generator().manual_seed(123)
perm = torch.randperm(len(train_idx_b), generator=g)
n_val = int(0.15 * len(train_idx_b))
val_idx_b = train_idx_b[perm[:n_val]]

model.eval()
with torch.no_grad():
    val_probs = torch.softmax(
        model(scada_data.x, scada_data.edge_index,
              scada_query_edges[:, val_idx_b], scada_edge_attr[val_idx_b]), dim=1
    )[:, 1]
val_y = scada_y[val_idx_b]

prec, rec, thresh = precision_recall_curve(val_y.numpy(), val_probs.numpy())
f1s = 2 * prec * rec / (prec + rec + 1e-9)
best_i = f1s.argmax()
chosen_threshold = thresh[best_i]
print(f"Chosen on VAL (held-out subset of train, ORIGINAL model, no retrain): "
      f"threshold={chosen_threshold:.3f}  precision={prec[best_i]:.4f}  recall={rec[best_i]:.4f}")

# Apply that fixed threshold ONCE, to the ORIGINAL model's test-set probabilities
# (test_probs / test_y were computed by this same `model` in the cell above --
# same weights, same test predictions used everywhere else in this notebook;
# only the decision boundary changes).
test_pred_calibrated = (test_probs >= chosen_threshold).long()

tp = ((test_pred_calibrated == 1) & (test_y == 1)).sum().item()
fp = ((test_pred_calibrated == 1) & (test_y == 0)).sum().item()
fn = ((test_pred_calibrated == 0) & (test_y == 1)).sum().item()
tn = ((test_pred_calibrated == 0) & (test_y == 0)).sum().item()
precision = tp / (tp + fp) if (tp + fp) else 0.0
recall = tp / (tp + fn) if (tp + fn) else 0.0
f1 = 2 * precision * recall / (precision + recall) if (precision + recall) else 0.0
fpr_normal = fp / (fp + tn) if (fp + tn) else 0.0
print(f"UNBIASED test result @ threshold={chosen_threshold:.3f} "
      f"(original model, isolated threshold effect only): "
      f"precision={precision:.4f}  recall={recall:.4f}  f1={f1:.4f}  FPR_normal={fpr_normal:.4f}")


# Explicit side-by-side: SAME test_probs, SAME test_y, only threshold changes
def eval_at_threshold(probs, y, t):
    pred = (probs >= t).long()
    tp_ = ((pred == 1) & (y == 1)).sum().item()
    fp_ = ((pred == 1) & (y == 0)).sum().item()
    fn_ = ((pred == 0) & (y == 1)).sum().item()
    tn_ = ((pred == 0) & (y == 0)).sum().item()
    p_ = tp_ / (tp_ + fp_) if (tp_ + fp_) else 0.0
    r_ = tp_ / (tp_ + fn_) if (tp_ + fn_) else 0.0
    f_ = 2 * p_ * r_ / (p_ + r_) if (p_ + r_) else 0.0
    fpr_ = fp_ / (fp_ + tn_) if (fp_ + tn_) else 0.0
    return dict(precision=p_, recall=r_, f1=f_, fpr=fpr_, tp=tp_, fp=fp_, fn=fn_, tn=tn_)

# Both computed from the exact same `test_probs` / `test_y` produced by `model`
# in cell 137 -- no recomputation, no retraining, nothing else changes.
result_default = eval_at_threshold(test_probs, test_y, 0.50)
result_calibrated = eval_at_threshold(test_probs, test_y, chosen_threshold)

print("=" * 78)
print(f"{'':30s}{'Track B @ 0.50 (default)':>24s}{'Track B @ calibrated':>24s}")
print("=" * 78)
print(f"{'Threshold':30s}{0.50:>24.3f}{chosen_threshold:>24.3f}")
print(f"{'Precision':30s}{result_default['precision']:>24.4f}{result_calibrated['precision']:>24.4f}")
print(f"{'Recall':30s}{result_default['recall']:>24.4f}{result_calibrated['recall']:>24.4f}")
print(f"{'F1':30s}{result_default['f1']:>24.4f}{result_calibrated['f1']:>24.4f}")
print(f"{'FPR (false positive rate)':30s}{result_default['fpr']:>24.4f}{result_calibrated['fpr']:>24.4f}")
print("=" * 78)
print(
    "Comparison performed using the same trained Track B model, "
    "the same test-set probabilities, and the same test labels; "
    "only the decision threshold changes."
)
````

### **Part J — Track B False-Positive Root Cause & Mitigation**

In Week 7, Track B showed a **27.2% false-positive rate (FPR)** on normal test traffic. This section investigates the root cause of these false positives and evaluates a simple threshold-calibration strategy to reduce them.

Before investigating the flow features, we first tested whether the false positives were caused by the topology/degree shortcut identified earlier. Node embeddings were analysed after replacing node features with constant vectors, and their variation remained minimal (standard deviation = **0.056**). This indicates that graph topology is **not** the primary cause of the false positives, suggesting that the model's decisions are driven mainly by the flow content.

#### **Root Cause**

The analysis showed that approximately **95% of all false positives** originated from only two hosts:

- `192.168.119.140` – 2,347 false positives
- `192.168.119.141` – 2,027 false positives

These hosts also generate most of the attack traffic in the dataset (e.g., **udp_flood**, **vuln_scan**, and **http_flood**). Although they also produce legitimate traffic, their normal flows exhibit characteristics that are statistically more similar to attack traffic than to normal traffic from other hosts.

For example:

- `Tcp_flags_reset_Set` appears in **44.1%** of false positives but only **0.1%** of true negatives.
- `ip_ttl` and `frame_len` also show clear distributional differences.

These observations suggest that the model is learning **host-specific traffic characteristics** rather than making random mistakes. Consequently, normal traffic generated by these dominant attack hosts is more likely to be classified as malicious.

#### **Mitigation**

To reduce false positives, the decision threshold was calibrated using a **validation subset** derived from the training data. The selected threshold (**0.790**) was then applied to the **same trained Track B model** and its existing test-set probabilities, ensuring that only the decision threshold changed during evaluation.

| Threshold | Precision | Recall | FPR |
|-----------|----------:|-------:|----:|
| Default (0.500) | 0.9508 | 0.9997 | 27.2% |
| Calibrated (0.790) | 0.9720 | 0.9974 | 15.2% |

The calibrated threshold reduced the false-positive rate from **27.2%** to **15.2%**, while reducing recall by only **0.23 percentage points**. ## Part K — GNN Explainability Setup

`EdgeClassifierWithAttr.forward` takes two extra required args beyond `(x, edge_index)` -- `query_edges` and `query_edge_attr` -- which explainer APIs don't know how to pass through. `FlowExplainWrapper` freezes one flow's `(query_edges, query_edge_attr)` as fixed attributes so the explainer can perturb `x` / `edge_index` (topology) or `edge_attr` (content) while the flow being explained stays fixed.

Two independent methods are applied to the topology question, per the roadmap item naming both:
- **GNNExplainer** -- per-instance optimization, no training phase.
- **PGExplainer** -- a parametric explainer network trained per flow. Two API   constraints in this PyG version shape the setup: PGExplainer only supports   `task_level='node'` or `'graph'` (not `'edge'`), so the single-flow output   is treated as a graph-level prediction; and it only supports   `explanation_type='phenomenon'`, so the model's *own predicted class* is   passed as the training target (not the true label) -- the goal is   explaining why the model predicted attack, not re-deriving the ground truth.

A third method, **Captum Integrated Gradients**, attributes each prediction to the 23 raw `edge_attr` content features directly -- neither GNNExplainer's nor PGExplainer's masks cover `edge_attr`, and content features are where Part J's qualitative analysis points to the actual signal.

````python
!pip install captum --quiet

from torch_geometric.explain import Explainer, GNNExplainer, PGExplainer
from captum.attr import IntegratedGradients
import matplotlib.pyplot as plt
````

````python
class FlowExplainWrapper(torch.nn.Module):
    """Freezes one flow's (query_edges, query_edge_attr) so an explainer
    can perturb x / edge_index without needing to know about the extra args.
    Output is a single row [1, num_classes] -- treated as a graph-level
    prediction for explainer API purposes (see task_level note above)."""
    def __init__(self, base_model, query_edges, query_edge_attr):
        super().__init__()
        self.base_model = base_model
        self.query_edges = query_edges
        self.query_edge_attr = query_edge_attr

    def forward(self, x, edge_index):
        return self.base_model(x, edge_index, self.query_edges, self.query_edge_attr)


def explain_flow_topology_gnn(model, data, src_idx, dst_idx, edge_attr_row, epochs=100):
    """GNNExplainer over (x, edge_index) for a single flow. Near-zero masks
    would formally confirm the cell-136 finding that topology carries no signal."""
    query_edges = torch.tensor([[src_idx], [dst_idx]])
    query_edge_attr = edge_attr_row.unsqueeze(0)
    wrapped = FlowExplainWrapper(model, query_edges, query_edge_attr)

    explainer = Explainer(
        model=wrapped,
        algorithm=GNNExplainer(epochs=epochs),
        explanation_type='model',
        node_mask_type='attributes',
        edge_mask_type='object',
        model_config=dict(mode='multiclass_classification',
                           task_level='graph', return_type='raw'),
    )
    return explainer(data.x, data.edge_index, index=0)


def explain_flow_topology_pg(model, data, src_idx, dst_idx, edge_attr_row, epochs=30, lr=0.003):
    """PGExplainer over (x, edge_index) for a single flow. Independent second
    method to cross-check the GNNExplainer topology result. Trains its own
    small edge-mask predictor for this flow (no cross-flow amortization,
    since each flow needs its own frozen wrapper)."""
    query_edges = torch.tensor([[src_idx], [dst_idx]])
    query_edge_attr = edge_attr_row.unsqueeze(0)
    wrapped = FlowExplainWrapper(model, query_edges, query_edge_attr)

    with torch.no_grad():
        predicted_class = wrapped(data.x, data.edge_index).argmax(dim=1)

    explainer = Explainer(
        model=wrapped,
        algorithm=PGExplainer(epochs=epochs, lr=lr),
        explanation_type='phenomenon',
        edge_mask_type='object',
        model_config=dict(mode='multiclass_classification',
                           task_level='graph', return_type='raw'),
    )
    for epoch in range(epochs):
        explainer.algorithm.train(epoch, wrapped, data.x, data.edge_index,
                                   target=predicted_class, index=0)
    return explainer(data.x, data.edge_index, target=predicted_class, index=0)


def explain_flow_features(model, data, query_edges, query_edge_attr, n_steps=50):
    """Integrated-Gradients attribution of P(attack) to each of the 23 raw
    edge_attr features, batched across many flows at once."""
    n_flows = query_edge_attr.shape[0]

    def forward_edge_attr(edge_attr_input):
        reps = edge_attr_input.shape[0] // n_flows
        qe = query_edges.repeat(1, reps)
        logits = model(data.x, data.edge_index, qe, edge_attr_input)
        return F.softmax(logits, dim=1)[:, 1]   # P(attack)

    ig = IntegratedGradients(forward_edge_attr)
    baseline = torch.zeros_like(query_edge_attr)
    return ig.attribute(query_edge_attr, baselines=baseline, n_steps=n_steps)

print("Explainability functions defined: GNNExplainer, PGExplainer, Captum IG")
````

# **Week 10 — Temporal / Streaming Graph Updates**

Weeks 2-9 all treat the network as a **static snapshot**: one graph is built
once from the *entire* dataset before any training happens, and the
evaluation splits used so far (random flow split in Week 4, IP-held-out
split in Weeks 6-9) don't respect **arrival order** — a training example can
sit "before" a test example in wall-clock time or "after" it, and the
topology (`edge_index`) is allowed to contain edges that, in a real
deployment, wouldn't exist yet.

This week reframes SCADANet as a **discrete-time dynamic graph (DTDG)**:
flows are sorted chronologically and chopped into ordered time windows, the
topology is built up **incrementally** window-by-window (streaming
semantics — a snapshot at window *w* may only use information from windows
`0..w`), and the model is trained on an early slice of time and evaluated on
a later slice it has never seen.

**Why SCADANet and not ICS-Flow:** ICS-Flow's 9-node graph is too small for
a meaningful time-windowed split (Week 4 already noted this for the
node-level task). SCADANet's `Time` column is a continuous per-flow
timestamp across 534,841 flows, which gives fine-grained temporal
resolution and lets this build directly on the Week 6-9 SCADANet /
`EdgeClassifierWithAttr` pipeline.

**Stated simplification:** the node ID space (`scada_ip_index`, `all_ips`)
is still taken from the *full* dataset up front, so this streams **edge**
arrival, not node arrival — a fully realistic streaming setting would also
need to handle previously-unseen IPs (cold-start nodes) appearing mid-capture.
That's called out explicitly as future work below rather than glossed over.

## Part L — Streaming Graph Construction

````python
# L.1 -- Chronological ordering & discrete-time windowing
# Keep the original row position (orig_idx) so predictions can be mapped
# back into the scada_query_edges / scada_edge_attr / scada_y tensors
# already built in Part G / G.2.
# Windows are equal-COUNT rather than equal-width-in-time: SCADANet's
# traffic is bursty (udp_flood dominates one region of the capture), so
# equal-width time bins put almost all flows in a handful of windows and
# left the rest nearly empty. Equal-count windows keep chronological order
# but guarantee every window (and therefore every split) has a comparable
# number of flows. The tradeoff is that window duration in seconds becomes
# uneven -- reported explicitly below rather than hidden, since it matters
# for how "temporal" the resulting train/test split actually is (see Part O).

N_WINDOWS = 20

scada_df_t = (
    scada_df.sort_values("Time")
    .reset_index()
    .rename(columns={"index": "orig_idx"})
)

t_min, t_max = scada_df_t["Time"].min(), scada_df_t["Time"].max()

scada_df_t["window"] = (np.arange(len(scada_df_t)) * N_WINDOWS) // len(scada_df_t)

window_counts = scada_df_t.groupby("window").size()
window_attack_rate = scada_df_t.groupby("window")["is_attack"].mean()
window_duration = scada_df_t.groupby("window")["Time"].agg(lambda s: s.max() - s.min())

print(f"Capture spans {t_max - t_min:.1f}s, split into {N_WINDOWS} equal-COUNT windows "
      f"(~{len(scada_df_t) // N_WINDOWS} flows each)")
print("\nFlows per window:")
print(window_counts)
print("\nAttack rate per window:")
print(window_attack_rate.round(3))
print("\nWall-clock duration per window (s) -- expect this UNEVEN, that's the point:")
print(window_duration.round(1))
````

````python
# L.2 -- Streaming/cumulative topology per window
# At window w, the topology graph = union of all IP pairs seen in windows
# 0..w (edges accumulate, mirroring how a KG is actually built in
# production -- once a communication edge is observed it stays known).
# Nothing from windows > w is allowed to leak into snapshot w.
def build_streaming_snapshots(df_t, ip_index, n_windows):
    snapshots = []
    seen_pairs = set()
    for w in range(n_windows):
        w_df = df_t[df_t["window"] == w]

        pairs = w_df[[SRC_IP_COL, DST_IP_COL]].astype(str).drop_duplicates()
        pairs = pairs[pairs[SRC_IP_COL].isin(ip_index) & pairs[DST_IP_COL].isin(ip_index)]
        new_pairs = set(zip(pairs[SRC_IP_COL], pairs[DST_IP_COL]))
        seen_pairs |= new_pairs

        if seen_pairs:
            src = [ip_index[a] for a, b in seen_pairs]
            dst = [ip_index[b] for a, b in seen_pairs]
            edge_index_w = torch.tensor(np.stack([src, dst]), dtype=torch.long)
        else:
            edge_index_w = torch.zeros((2, 0), dtype=torch.long)

        snapshots.append({
            "window": w,
            "n_edges_seen": len(seen_pairs),
            "n_flows_in_window": len(w_df),
            "edge_index": edge_index_w,
        })
    return snapshots

streaming_snapshots = build_streaming_snapshots(scada_df_t, scada_ip_index, N_WINDOWS)

for s in streaming_snapshots:
    print(f"window {s['window']:>2}: "
          f"{s['n_flows_in_window']:>6} new flows | "
          f"cumulative topology edges={s['n_edges_seen']:>4}")
````

## Part M — Temporal Train/Test Split

Contrast with the Week 6-9 IP-held-out split: that split holds out *IPs*,
so the same time period appears on both sides of train/test and the
question it answers is "does the model generalize to unseen hosts". A
**temporal split** instead trains on windows `[0, SPLIT_WINDOW)` and tests
on `[SPLIT_WINDOW, N_WINDOWS)` -- the question it answers is "does the
model trained on the graph's early behavior still work later", which is
the realistic deployment question and is sensitive to **concept drift**
(e.g. attack types that only start appearing partway through the capture).

Because IP identity stays available on both sides of a temporal split
(unlike the IP-held-out split), this also isolates the "does the model rot
over time" question from the "does the model just memorize IPs" question
Week 6 already answered -- the two splits are diagnosing different failure
modes, not competing for the same title of "the" evaluation.

````python
# M.1 -- Temporal (time-ordered) train/test split
SPLIT_WINDOW = int(N_WINDOWS * 0.7)  # 70% of flows by chronological order; not 70% of wall-clock time

train_mask_t = (scada_df_t["window"] < SPLIT_WINDOW).to_numpy()
test_mask_t = (scada_df_t["window"] >= SPLIT_WINDOW).to_numpy()
print(f"Split at window {SPLIT_WINDOW}: "
      f"{train_mask_t.sum()} train flows, {test_mask_t.sum()} test flows")
print("Train-window attack rate:", scada_df_t.loc[train_mask_t, "is_attack"].mean().round(3))
print("Test-window attack rate:", scada_df_t.loc[test_mask_t, "is_attack"].mean().round(3))

# Attack types present in test but never seen in train = a concept-drift
# check -- if this list is non-empty, low recall on those rows later is
# expected and shouldn't be read as a general model failure.
train_types = set(scada_df_t.loc[train_mask_t, LABEL_COL].unique())
test_types = set(scada_df_t.loc[test_mask_t, LABEL_COL].unique())
unseen_in_train = test_types - train_types
print("Attack types in test but never seen during training:", unseen_in_train or "none")

# Topology for TRAINING must reflect only what had been seen by the end of
# the training period -- pull it straight from the streaming snapshot
# instead of rebuilding it from the full (train+test) dataframe.
train_topology_edge_index = streaming_snapshots[SPLIT_WINDOW - 1]["edge_index"]

# The graph keeps growing during the test period too (new IP *pairs*, not
# new IPs -- see the stated simplification above).
#
# IMPORTANT: this evaluation uses the FINAL cumulative test-period topology
# for all test flows. Therefore, this is not strict flow-by-flow online
# inference: later test-period edges are available to earlier test examples.
# A strict online evaluation would evaluate each test window using only the
# topology accumulated up to that window
full_topology_edge_index = streaming_snapshots[-1]["edge_index"]
new_pairs_in_test = (streaming_snapshots[-1]["n_edges_seen"]
                      - streaming_snapshots[SPLIT_WINDOW - 1]["n_edges_seen"])
print(f"New IP-pair edges first observed during the test period: {new_pairs_in_test}")
````

### Temporal split interpretation

This is a **70/30 chronological split by flow count**, not by wall-clock duration. Because SCADANet traffic is highly bursty, equal-count windows have very different time durations. In this experiment, the training period covers only about 49.9 seconds, while the test period covers about 1998.5 seconds.

The split therefore represents a challenging temporal distribution-shift evaluation rather than a balanced 70/30 split in real time.

````python
# M.2 -- Train on the early window, evaluate on the later window
def train_eval_scadanet_temporal(train_edge_index, eval_edge_index,
                                  query_edges, edge_attr, y,
                                  train_pos, test_pos,
                                  n_nodes, epochs=400, lr=0.003, grad_clip=1.0):
    x = torch.ones((n_nodes, 4), dtype=torch.float32)
    train_data = Data(x=x, edge_index=train_edge_index)
    eval_data = Data(x=x, edge_index=eval_edge_index)

    train_idx = torch.tensor(train_pos, dtype=torch.long)
    test_idx = torch.tensor(test_pos, dtype=torch.long)
    train_y, test_y = y[train_idx], y[test_idx]

    class_counts = torch.bincount(train_y, minlength=2).float()
    class_weights = train_y.shape[0] / (2.0 * class_counts.clamp(min=1))
    print(f"Class counts (train window): normal={int(class_counts[0])}, "
          f"attack={int(class_counts[1])} -> weights={class_weights.tolist()}")

    model = EdgeClassifierWithAttr(in_dim=x.shape[1], edge_attr_dim=edge_attr.shape[1])
    optimizer = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=5e-4)

    model.train()
    for epoch in range(epochs):
        optimizer.zero_grad()
        out = model(train_data.x, train_data.edge_index,
                    query_edges[:, train_idx], edge_attr[train_idx])
        loss = F.cross_entropy(out, train_y, weight=class_weights)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=grad_clip)
        optimizer.step()
        if (epoch + 1) % 100 == 0:
            print(f"  epoch {epoch + 1}: loss={loss.item():.4f}")

    model.eval()
    with torch.no_grad():
        test_pred = model(eval_data.x, eval_data.edge_index,
                           query_edges[:, test_idx], edge_attr[test_idx]).argmax(dim=1)
        acc = (test_pred == test_y).float().mean().item()
        tp = ((test_pred == 1) & (test_y == 1)).sum().item()
        fp = ((test_pred == 1) & (test_y == 0)).sum().item()
        fn = ((test_pred == 0) & (test_y == 1)).sum().item()
        tn = ((test_pred == 0) & (test_y == 0)).sum().item()
        precision = tp / (tp + fp) if (tp + fp) else 0.0
        recall = tp / (tp + fn) if (tp + fn) else 0.0
        f1 = 2 * precision * recall / (precision + recall) if (precision + recall) else 0.0

    return model, {
        "test_accuracy": acc, "precision": precision, "recall": recall, "f1": f1,
        "n_train_flows": len(train_idx), "n_test_flows": len(test_idx),
        "confusion_matrix": {"tp": tp, "fp": fp, "fn": fn, "tn": tn},
    }

train_pos = scada_df_t.loc[train_mask_t, "orig_idx"].to_numpy()
test_pos = scada_df_t.loc[test_mask_t, "orig_idx"].to_numpy()

temporal_model, temporal_metrics = train_eval_scadanet_temporal(
    train_topology_edge_index, full_topology_edge_index,
    scada_query_edges, scada_edge_attr, scada_y,
    train_pos, test_pos, n_nodes=len(all_ips),
)
print("\nTemporal-split results:", temporal_metrics)
````

## **Part O — Findings**

### 1. Graph growth is back-loaded, not front-loaded

Cumulative topology edges remain relatively small during the early windows,
increasing from **8 edges in window 0 to 536 edges by window 12**. The topology
then grows sharply across windows 13–16, increasing from **820 to 2479, 4772,
and 6346 edges**, before approaching a plateau at **6535 edges by window 19**.

This shows that most IP-pair topology is discovered during the later part of
the capture rather than during the initial period. Consequently, the training
topology snapshot (`train_topology_edge_index`), taken at the end of window
13, contains only a small subset of the eventual graph: **820 cumulative
edges at the training boundary versus 6535 edges by the end of the capture**.
A total of **5715 IP-pair edges first appear during the test period**.

### 2. Temporal split produces a substantial traffic-distribution shift

The temporal split produces a substantial change in the traffic distribution.
The training period has an attack rate of **98.2%**, whereas the test period
has an attack rate of **51.9%**. The shift becomes particularly pronounced in
the final windows, where the attack rate falls to **3.9% in window 18** and
**0.6% in window 19**.

Therefore, the Week 10 experiment evaluates not only generalization to later
traffic in chronological order, but also generalization across a substantial
change in the attack/normal traffic distribution.

### 3. Temporal attack-distribution shift is substantial

The chronological attack-type distribution changes substantially across the
20 windows. The early portion of the capture is dominated by `udp_flood`,
which accounts for 98.10% of window 0 and remains above 84% in several early
windows. This composition changes sharply around windows 13-17: `udp_flood`
falls to 62.39% in window 13 and disappears completely from windows 14 onward,
while `http_flood`, `icmp_flood`, `normal_traffic`, and especially `vuln_scan`
become increasingly prominent.

The shift is particularly strong in windows 15-17, where `vuln_scan` accounts
for 31.15%, 44.18%, and 48.18% of the respective windows. In contrast, the
final windows become overwhelmingly normal traffic, with `normal_traffic`
representing 96.06% of window 18 and 99.44% of window 19.

This provides direct evidence that the temporal test period is not simply a
later sample from the same traffic distribution: its attack-type composition
is substantially different from the early training period. This temporal
distribution shift is therefore an important context for interpreting the
model's temporal generalization performance.

### 4. Recall degradation is concentrated in specific attack types

The temporal performance degradation is not uniform across attack classes.
`vuln_scan` is the dominant source of false negatives: it has n=37,777 test
flows and recall=0.0084, accounting for approximately 88.6% of all false
negatives. `modbus_fdi` (n=2,636), `modbus_abuse` (n=498), and
`insider_threat` (n=1,446) also have recall=0.0.

In contrast, several attack types remain highly detectable:
`icmp_flood`, `http_flood`, `tcp_ack_flood`, `sql_injection`, and `apt_exfil`
all achieve recall=1.0, while `tcp_syn_flood` and `port_scan` achieve recall
of 0.9687 and 0.9495 respectively. *(Updated to match the current run's
`week10_temporal_streaming.per_attack_type_breakdown`: `tcp_syn_flood` recall
0.9673->0.9687 and `port_scan` recall corrected from a stale 0.9560 to
0.9495/0.9488 -- the 0.9560 figure did not match either the old or the new
run and appears to have been a pre-existing transcription error, not
something introduced by this rerun.)*

Importantly, no attack type in the temporal test set is completely unseen
during training (`unseen_attack_types_in_test` is empty). Therefore, the
degradation is not explained by the appearance of entirely new attack
classes. Instead, the results indicate difficulty generalizing within
several already-known attack classes under the changed temporal traffic
distribution.

### 5. The 70/30 flow-count split produces a large wall-clock imbalance

The training window (windows 0-13) spans only **~49.9s** of the 2048.6s
capture, while the test window (windows 14-19) spans **~1998.5s**. Thus,
although the split is approximately 70/30 by flow count, it is approximately
2.4%/97.6% by wall-clock duration.

This occurs because the traffic is highly bursty: equal numbers of flows can
occur over very different amounts of time. Therefore, this experiment should
be described as a chronological equal-flow-count split, not as a 70/30
time-duration split.

This is a limitation of the current experimental design and motivates a
duration-aware temporal windowing experiment as possible future work.

### 6. Limitations

- **Fixed node ID space:** The node ID space (`scada_ip_index`, `all_ips`) is constructed from the full dataset before streaming. Consequently, this experiment models **edge arrival rather than node arrival**. A previously unseen IP appearing during the capture is therefore not treated as a true cold-start node.

- **Not true online learning:** The experiment uses discretized time windows and cumulative topology updates, but the model is trained once on the complete chronological training portion. It is **not updated flow-by-flow or at each test-window boundary**. True incremental learning, such as per-flow SGD with a replay buffer to address catastrophic forgetting, is outside the scope of Week 10.

- **Final cumulative topology during test evaluation:** The current evaluation uses the final cumulative topology for the test-period predictions. Therefore, it should not be interpreted as strict flow-by-flow online inference, because later test-period edges are available to earlier test examples. A strict online evaluation would evaluate each test window using only the topology accumulated up to that window.

- **Uneven wall-clock durations:** Equal-count windowing ensures a comparable number of flows per window, but it does not produce comparable time durations. As shown above, the training and test periods differ substantially in wall-clock coverage. A future experiment could investigate duration-aware windowing as an alternative.

- **Interpretation of feature drift:** The Week 9 explainability results identified `Protocol_TCP`, `Tcp_flags_reset_Set`, and `frame_len` as important content-feature drivers. The Week 10 results motivate the hypothesis that content-feature distributions may also change over time for the affected attack classes. However, Week 10 does not directly measure feature distributions across windows and therefore does not independently establish feature drift or isolate it from topology evolution. This should be treated as a hypothesis for future analysis rather than a demonstrated causal explanation.

## Part R — Label Permutation Test: Is BATADAL Temporal recall=1.0 Real?

Perfect recall on a held-out temporal split isn't automatically memorization
(the model never saw these windows during training), but it's worth testing
directly rather than assuming. The standard check: shuffle the labels so any
true signal between features/topology and attack/normal is destroyed, retrain
under the exact same split, and see what the model achieves. If shuffled-label
performance stays high, something is leaking through the features or topology
independent of the labels. If it collapses toward chance, the real result
reflects a genuinely learnable (if easy) signal, not leakage.

````python
# R.1 -- Label permutation test on BATADAL temporal split
# Same split boundary (batadal_split_idx) and same window features/topology as
# P.1 -- only the labels are randomized, per graph, before splitting.

import copy

N_PERMUTATIONS = 5
permutation_results = []

for seed in range(N_PERMUTATIONS):
    rng = np.random.RandomState(seed)
    shuffled_labels = rng.permutation(window_labels)

    shuffled_graphs = []
    for g, lbl in zip(window_graphs, shuffled_labels):
        g2 = copy.deepcopy(g)
        g2.y = torch.tensor([int(lbl)], dtype=torch.long)
        shuffled_graphs.append(g2)

    train_shuf = shuffled_graphs[:batadal_split_idx]
    test_shuf = shuffled_graphs[batadal_split_idx:]

    _, shuf_metrics = run_fold(train_shuf, test_shuf, seed=seed)
    permutation_results.append(shuf_metrics)
    print(f"Seed {seed}: acc={shuf_metrics['accuracy']:.4f} "
          f"prec={shuf_metrics['precision']:.4f} rec={shuf_metrics['recall']:.4f} "
          f"f1={shuf_metrics['f1']:.4f}")

perm_precs = np.array([r['precision'] for r in permutation_results])
perm_recs = np.array([r['recall'] for r in permutation_results])
perm_f1s = np.array([r['f1'] for r in permutation_results])

print("\n" + "=" * 60)
print("PERMUTATION TEST -- REAL vs. SHUFFLED LABELS (BATADAL temporal)")
print("=" * 60)
print(f"{'Metric':<12}{'Real labels':<16}{'Shuffled (mean±std)':<24}")
print(f"{'Precision':<12}{batadal_temporal_metrics['precision']:<16.4f}"
      f"{perm_precs.mean():.4f} ± {perm_precs.std():.4f}")
print(f"{'Recall':<12}{batadal_temporal_metrics['recall']:<16.4f}"
      f"{perm_recs.mean():.4f} ± {perm_recs.std():.4f}")
print(f"{'F1':<12}{batadal_temporal_metrics['f1']:<16.4f}"
      f"{perm_f1s.mean():.4f} ± {perm_f1s.std():.4f}")

leakage_flag = perm_f1s.mean() > 0.5 * batadal_temporal_metrics['f1']
print(f"\nShuffled F1 is {'>50%' if leakage_flag else '<50%'} of real F1 -- "
      f"{'INVESTIGATE possible leakage' if leakage_flag else 'consistent with a real, learnable signal'}")
````

## Part S — Split-Boundary Sensitivity Check

Part R confirmed the temporal result isn't memorization. This checks a
different failure mode: whether recall=1.0 is specific to the 70/30 boundary
(21 positive test windows, possibly a lucky cut) or holds as the split moves.
Runs the same chronological split at 60/40, 70/30, and 80/20 and reports
metrics at each, so "the model generalizes over time" rests on more than one
split.

````python
# S.1 -- Rolling split-boundary sensitivity check on BATADAL temporal split
# Same window_graphs / window_labels (already time-ordered), same run_fold.
# Only the split fraction changes -- topology/features are untouched.

SPLIT_FRACTIONS = [0.6, 0.7, 0.8]
sensitivity_results = []

for frac in SPLIT_FRACTIONS:
    split_idx = int(len(window_graphs) * frac)
    train_s = window_graphs[:split_idx]
    test_s = window_graphs[split_idx:]
    n_test_pos = int(window_labels[split_idx:].sum())

    _, metrics_s = run_fold(train_s, test_s, seed=42)
    sensitivity_results.append({
        "split_frac": frac,
        "split_idx": split_idx,
        "n_train": len(train_s),
        "n_test": len(test_s),
        "n_test_positive": n_test_pos,
        **metrics_s,
    })
    print(f"Split {int(frac*100)}/{int((1-frac)*100)}: "
          f"train={len(train_s)} test={len(test_s)} (pos={n_test_pos}) | "
          f"prec={metrics_s['precision']:.4f} rec={metrics_s['recall']:.4f} "
          f"f1={metrics_s['f1']:.4f}")

print("\n" + "=" * 70)
print("SPLIT-BOUNDARY SENSITIVITY -- BATADAL TEMPORAL")
print("=" * 70)
print(f"{'Split':<10}{'Test windows':<14}{'Test pos':<10}{'Precision':<12}{'Recall':<12}{'F1':<12}")
for r in sensitivity_results:
    print(f"{int(r['split_frac']*100)}/{int((1-r['split_frac'])*100):<7}"
          f"{r['n_test']:<14}{r['n_test_positive']:<10}"
          f"{r['precision']:<12.4f}{r['recall']:<12.4f}{r['f1']:<12.4f}")

recalls_all = [r['recall'] for r in sensitivity_results]
recall_stable = min(recalls_all) >= 0.9
print(f"\nRecall stays >=0.9 across all splits: {recall_stable}")
print("-> " + ("recall=1.0 at 70/30 is not a lucky cut; the signal holds across boundaries."
      if recall_stable else
      "recall varies meaningfully with split boundary -- report per-split, not a single number."))
````

## Part T — Precision-Recall Curve (BATADAL Temporal Split)

Parts R and S establish that recall=1.0 at the default 0.5 threshold is a
real, stable signal, not leakage or a lucky split boundary. This section
shows *why* it lands at exactly 1.0: the class-weighted loss pushes the
decision boundary aggressively toward the minority (attack) class, so recall
sits at its ceiling while precision absorbs the cost. Plotting the full
precision-recall curve (same approach as the Week 8 Track B threshold
calibration) shows the whole trade-off rather than one threshold's point
estimate.

````python
# T.1 -- Precision-Recall curve for BATADAL temporal split (70/30)
# Retrains at the same 70/30 boundary as P.1, but keeps the model and raw
# softmax probabilities instead of only the argmax prediction.

from sklearn.metrics import precision_recall_curve, average_precision_score
from torch_geometric.loader import DataLoader

def run_fold_with_probs(train_graphs, test_graphs, epochs=100, seed=42):
    torch.manual_seed(seed)
    train_loader = DataLoader(train_graphs, batch_size=16, shuffle=True)
    test_loader = DataLoader(test_graphs, batch_size=32, shuffle=False)

    train_y = torch.cat([g.y for g in train_graphs])
    class_counts = torch.bincount(train_y, minlength=2).float()
    class_weights = train_y.shape[0] / (2.0 * class_counts.clamp(min=1))

    model = WindowGraphClassifier(in_dim=train_graphs[0].x.shape[1])
    optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=5e-4)

    model.train()
    for epoch in range(epochs):
        for batch in train_loader:
            optimizer.zero_grad()
            out = model(batch.x, batch.edge_index, batch.batch)
            loss = F.cross_entropy(out, batch.y, weight=class_weights)
            loss.backward()
            optimizer.step()

    model.eval()
    all_probs, all_true = [], []
    with torch.no_grad():
        for batch in test_loader:
            out = model(batch.x, batch.edge_index, batch.batch)
            probs = F.softmax(out, dim=1)[:, 1]
            all_probs.append(probs)
            all_true.append(batch.y)
    probs = torch.cat(all_probs).numpy()
    true = torch.cat(all_true).numpy()
    return model, probs, true


_, test_probs, test_true = run_fold_with_probs(
    train_graphs_temporal, test_graphs_temporal, seed=42
)

precisions, recalls, thresholds = precision_recall_curve(test_true, test_probs)
ap_score = average_precision_score(test_true, test_probs)

fig, ax = plt.subplots(figsize=(7, 6))
ax.plot(recalls, precisions, marker=".", color="#4C72B0", label=f"PR curve (AP={ap_score:.4f})")
ax.scatter([batadal_temporal_metrics["recall"]], [batadal_temporal_metrics["precision"]],
           color="#DD8452", s=80, zorder=5, label="Default threshold (0.5)")
ax.set_xlabel("Recall")
ax.set_ylabel("Precision")
ax.set_title("BATADAL Temporal Split -- Precision-Recall Curve")
ax.set_xlim(0, 1.02)
ax.set_ylim(0, 1.02)
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("week11_batadal_temporal_pr_curve.png", dpi=150)
plt.show()

print(f"Average Precision (AP): {ap_score:.4f}")
print(f"Precision at recall=1.0 (default threshold 0.5): {batadal_temporal_metrics['precision']:.4f}")

print("\nPrecision at selected recall levels:")
for target_recall in [0.7, 0.8, 0.9, 1.0]:
    idx = int(np.argmin(np.abs(recalls - target_recall)))
    thresh_str = f"{thresholds[idx]:.3f}" if idx < len(thresholds) else "min (0.0)"
    print(f"  recall~{recalls[idx]:.3f}  ->  precision={precisions[idx]:.3f}  (threshold~{thresh_str})")
````

## Part U — Threshold Calibration (BATADAL Temporal, Leave-One-Fold-Out)

**Fix (Week 11 review):** the original version of this section selected the
threshold directly off the Part T precision-recall curve, which was computed
on the TEST set — the same test-set leakage pattern already caught and fixed
for SCADANet Track B in Week 8 (Part J), but not reapplied here. This is now
corrected.

The threshold is selected using only `train_graphs_temporal` (485 windows,
35 positive), never `test_graphs_temporal`. Because the temporal train set is
small and the positive class is rare (7.2%), a single train/validation split
would leave very few positive validation windows, so a 5-fold stratified
**leave-one-fold-out** procedure is used instead: 5 separate models are each
trained on 4/5 of the training windows and evaluated on the held-out 1/5.
Stacking these out-of-fold (OOF) predictions gives every training window a
prediction from a model that never saw it during training — a proxy for
"unseen data" for threshold selection, without touching the test set at all.

The threshold chosen on the OOF predictions is then applied **once** to
`test_probs` / `test_true` — the fixed predictions from T.1, produced by the
single model trained on the full `train_graphs_temporal` set. Only the
decision boundary changes; the evaluated model and its test predictions are
exactly the ones used everywhere else in this notebook (P.1, S.1, T.1),
matching the isolation pattern used for the Week 8 Track B calibration (same
trained model, same test predictions, threshold-only comparison).

````python
# U.1 -- Leave-one-fold-out (5-fold) threshold calibration on TRAIN only
# Fixes the test-set leakage flagged in review: Part U previously picked the
# threshold off the Part T PR curve, which was computed on test_probs/test_true.
# Here, only train_graphs_temporal / train_labels_temporal are used -- the
# test set is not touched anywhere in this cell.

from sklearn.model_selection import StratifiedKFold

N_FOLDS = 5
skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=42)

oof_probs = np.zeros(len(train_graphs_temporal))
oof_true = np.array(train_labels_temporal)

for fold_i, (fit_idx, holdout_idx) in enumerate(
        skf.split(np.zeros(len(train_graphs_temporal)), train_labels_temporal)):
    fit_graphs = [train_graphs_temporal[i] for i in fit_idx]
    holdout_graphs = [train_graphs_temporal[i] for i in holdout_idx]

    # Same architecture/training procedure as T.1 (run_fold_with_probs), just
    # trained on 4/5 of the training windows instead of all of them.
    fold_model, _, _ = run_fold_with_probs(fit_graphs, holdout_graphs, epochs=100, seed=42)

    fold_loader = DataLoader(holdout_graphs, batch_size=32, shuffle=False)
    fold_model.eval()
    fold_probs = []
    with torch.no_grad():
        for batch in fold_loader:
            out = fold_model(batch.x, batch.edge_index, batch.batch)
            fold_probs.append(F.softmax(out, dim=1)[:, 1])
    fold_probs = torch.cat(fold_probs).numpy()

    oof_probs[holdout_idx] = fold_probs
    print(f"Fold {fold_i + 1}/{N_FOLDS}: {len(holdout_idx)} windows, "
          f"{int(oof_true[holdout_idx].sum())} positive")

print(f"\nOOF predictions collected for all {len(oof_probs)} training windows "
      f"({int(oof_true.sum())} positive). Test set untouched.")

oof_precisions, oof_recalls, oof_thresholds = precision_recall_curve(oof_true, oof_probs)
oof_f1_scores = np.where(
    (oof_precisions + oof_recalls) > 0,
    2 * oof_precisions * oof_recalls / (oof_precisions + oof_recalls + 1e-12),
    0.0,
)[:-1]

best_f1_idx_oof = int(np.argmax(oof_f1_scores))
best_f1_threshold_oof = float(oof_thresholds[best_f1_idx_oof])

precision_floor = 0.8
above_floor_oof = np.where(oof_precisions[:-1] >= precision_floor)[0]
if len(above_floor_oof) > 0:
    floor_idx_oof = above_floor_oof[np.argmax(oof_recalls[:-1][above_floor_oof])]
    floor_threshold_oof = float(oof_thresholds[floor_idx_oof])
else:
    floor_idx_oof, floor_threshold_oof = None, None

print(f"\nOOF-selected thresholds (train-only, no test labels used):")
print(f"  Max F1               : {best_f1_threshold_oof:.3f}  "
      f"(OOF precision={oof_precisions[best_f1_idx_oof]:.3f}, recall={oof_recalls[best_f1_idx_oof]:.3f})")
if floor_threshold_oof is not None:
    print(f"  Precision>=0.8 floor : {floor_threshold_oof:.3f}  "
          f"(OOF precision={oof_precisions[floor_idx_oof]:.3f}, recall={oof_recalls[floor_idx_oof]:.3f})")
else:
    print(f"  Precision>=0.8 floor : not reached on OOF predictions")
````

````python
# U.2 -- Apply OOF-selected thresholds ONCE to the held-out TEST set
# test_probs / test_true are unchanged from T.1: predictions from the single
# model trained on all of train_graphs_temporal. No retraining happens here,
# and no test label was used to choose either threshold in U.1.

def confusion_at_threshold(probs, true, thresh):
    pred = (probs >= thresh).astype(int)
    tp = int(((pred == 1) & (true == 1)).sum())
    fp = int(((pred == 1) & (true == 0)).sum())
    fn = int(((pred == 0) & (true == 1)).sum())
    tn = int(((pred == 0) & (true == 0)).sum())
    prec = tp / (tp + fp) if (tp + fp) else 0.0
    rec = tp / (tp + fn) if (tp + fn) else 0.0
    f1 = 2 * prec * rec / (prec + rec) if (prec + rec) else 0.0
    return {"threshold": float(thresh), "precision": prec, "recall": rec, "f1": f1,
            "confusion_matrix": {"tp": tp, "fp": fp, "fn": fn, "tn": tn}}

default_result = confusion_at_threshold(test_probs, test_true, 0.5)
best_f1_result = confusion_at_threshold(test_probs, test_true, best_f1_threshold_oof)
floor_result = (confusion_at_threshold(test_probs, test_true, floor_threshold_oof)
                if floor_threshold_oof is not None else None)

print("=" * 78)
print("THRESHOLD COMPARISON -- BATADAL TEMPORAL SPLIT (leave-one-fold-out calibration)")
print("=" * 78)
print(f"{'Operating point':<28}{'Thresh':<10}{'Prec':<8}{'Recall':<8}{'F1':<8}{'CM (tp/fp/fn/tn)'}")
r = default_result
print(f"{'Default (0.5)':<28}{r['threshold']:<10.3f}{r['precision']:<8.3f}"
      f"{r['recall']:<8.3f}{r['f1']:<8.3f}{r['confusion_matrix']}")
r = best_f1_result
print(f"{'Max F1 (OOF-selected)':<28}{r['threshold']:<10.3f}{r['precision']:<8.3f}"
      f"{r['recall']:<8.3f}{r['f1']:<8.3f}{r['confusion_matrix']}")
if floor_result is not None:
    r = floor_result
    print(f"{'Precision>=0.8 (OOF)':<28}{r['threshold']:<10.3f}{r['precision']:<8.3f}"
          f"{r['recall']:<8.3f}{r['f1']:<8.3f}{r['confusion_matrix']}")
else:
    print(f"{'Precision>=0.8 (OOF)':<28}not reached")

if floor_result is not None:
    print(f"\nF1 vs default: max-F1 {best_f1_result['f1']:.4f} "
          f"({(best_f1_result['f1'] - default_result['f1'])*100:+.1f} pts), "
          f"precision-floor {floor_result['f1']:.4f} "
          f"({(floor_result['f1'] - default_result['f1'])*100:+.1f} pts)")
else:
    print(f"\nF1 vs default: max-F1 {best_f1_result['f1']:.4f} "
          f"({(best_f1_result['f1'] - default_result['f1'])*100:+.1f} pts), "
          f"precision floor not reached")
````

## **Part V — Findings (Parts P-U)**

### 1. Temporal evaluation affects SCADANet more strongly than BATADAL

| Dataset  | Static F1 | Temporal F1 | Drop |
|---|---:|---:|---:|
| SCADANet | 0.9926 | 0.6286 | **-36.4 pts** |
| BATADAL  | 0.8621 | 0.7778 | **-8.4 pts** |

SCADANet's static protocol is known from Weeks 6-8 to be structurally easy. Ten of the 14 attack types use a single fixed source IP and are absent from the IP-held-out test set. As a result, the static recall of 1.0 appears to reflect the relative ease of detecting the two spoofed-IP attack types rather than broad generalization across attack types.

The temporal split removes this shortcut, and recall decreases to 0.4927, consistent with the Week 10 finding that `vuln_scan` accounts for approximately 88.6% of temporal false negatives. This suggests that the temporal result provides a more challenging estimate of the model's ability to generalize to later attack patterns. *(Recall and F1 updated from 0.4935/0.6292 to 0.4927/0.6286 to match the current `week11_static_vs_temporal_benchmark.scadanet.temporal` run -- a small run-to-run shift of the same non-deterministic-training kind seen elsewhere in Week 10/11, not related to the seed-7 or PSI bugs.)*

---

### 2. The two datasets show different temporal failure patterns

- **SCADANet (temporal):** recall decreases substantially (1.0 → 0.49), while precision remains relatively high (0.99 → 0.87). The model misses a larger proportion of attacks in the later test period, particularly attack patterns such as `vuln_scan` that do not generalize well temporally.

- **BATADAL (temporal):** recall remains at 1.0000 at the default threshold (0.9652 → 1.0000), while precision decreases (0.79 → 0.64). The model continues to detect all attack windows in the later test period, but produces more false positives.

For the 70/30 BATADAL temporal split, the default threshold produces:

- True positives (TP): **21**
- False positives (FP): **12**
- False negatives (FN): **0**
- True negatives (TN): **175**

Therefore, recall is 1.0 because all 21 attack windows are detected, but this comes at the cost of 12 false positives.

---

### 3. BATADAL recall=1.0 is supported by the diagnostic checks

Three checks were performed to examine whether the perfect recall could be explained by memorization, label-independent artifacts, or the specific choice of the 70/30 split boundary.

- **Label permutation test (Part R):** shuffling `window_labels` across 5 seeds and retraining under the identical split reduces F1 from **0.7778** to **0.0384 ± 0.0545**, giving an approximately 20x gap. The large performance drop after label randomization is consistent with the model learning information associated with the real attack labels rather than relying only on label-independent structure.

- **Split-boundary sensitivity (Part S):** re-running the chronological split at **60/40, 70/30, and 80/20** keeps recall at **1.0000** across all three boundaries. This suggests that the recall result is not specific to the particular 70/30 split.

- **Conclusion:** together, these checks provide evidence that the model is learning a meaningful signal from sensor z-score deviations against the fixed `G_normal` topology, rather than simply memorizing the training windows or relying on one particular split boundary.

---

### 4. Recall=1.0 appears to be strongly influenced by the decision threshold

The Part T precision-recall analysis showed that the model can achieve high recall, but precision decreases as recall approaches 1.0.

The default classification threshold is **0.5**. Because `run_fold` uses a class-weighted loss, the minority attack class receives substantially greater weight. This encourages the model to avoid false negatives and can therefore make the model more willing to classify a window as an attack.

This explains the observed operating point:

| Threshold | Precision | Recall | F1 |
|---:|---:|---:|---:|
| 0.5 | 0.636 | 1.000 | 0.778 |

Therefore, recall=1.0 should **not** be interpreted as perfect classification.

The model is detecting every attack window, but it is also incorrectly classifying 12 normal windows as attacks.

---

### 5. Threshold calibration was corrected to avoid test-set leakage

The original threshold calibration selected the threshold directly from the **test-set precision-recall curve**. This was a test-set leakage issue because the test labels were being used to choose the operating threshold.

Following the Week 11 review feedback, this was corrected using a **5-fold leave-one-fold-out procedure on the training set only**.

The temporal training set contains:

- **485 training windows**
- **35 positive windows**
- **450 normal windows**

Five models were trained using four folds for training and one fold for generating out-of-fold (OOF) predictions.

Each training window therefore receives a prediction from a model that did not train on that particular window.

The threshold was then selected using these OOF predictions.

The test set was not used during threshold selection.

The OOF procedure produced predictions for all **485 training windows**, with **35 positive windows**.

The test set contains **208 windows**, including **21 positive windows**.

Most importantly, **test labels were not used to select the threshold**.

---

### 6. Corrected threshold calibration results

The OOF procedure selected two possible operating points:

| Operating point | Threshold | OOF Precision | OOF Recall |
|---|---:|---:|---:|
| Max F1 | 0.968 | 0.968 | 0.857 |
| Precision ≥ 0.8 | 0.667 | 0.800 | 0.914 |

These thresholds were then applied **once** to the fixed BATADAL temporal test predictions from Part T.

| Operating point | Threshold | Precision | Recall | F1 | Confusion (TP/FP/FN/TN) |
|---|---:|---:|---:|---:|---|
| Default (0.5) | 0.500 | 0.636 | 1.000 | 0.778 | 21/12/0/175 |
| Max F1 (OOF-selected) | 0.968 | 0.677 | 1.000 | 0.808 | 21/10/0/177 |
| Precision ≥ 0.8 (OOF) | 0.667 | 0.636 | 1.000 | 0.778 | 21/12/0/175 |

The **Max-F1 OOF-selected threshold of 0.968** improves the test F1 from:

- Default F1: **0.7778**
- Calibrated F1: **0.8077**
- Improvement: **+3.0 points**

It also reduces false positives from **12 to 10** while still detecting all **21 attack windows**.

The precision-floor threshold of **0.667** did not improve the test result over the default threshold, even though it achieved the desired precision level on the OOF training predictions.

This demonstrates why the threshold must be selected without looking at the test labels: the OOF estimate does not guarantee the same operating point on future test data.

The corrected calibration therefore gives a more conservative and leakage-free estimate of the benefit of threshold adjustment. The improvement is **modest (+3.0 F1 points)** compared with the earlier test-set-based calibration.

---

### 7. Caveats

- **Small positive test count:** 21 positive windows are present in the 70/30 temporal test set. Therefore, precision, recall, and F1 are based on a relatively small number of positive examples and should be interpreted with appropriate uncertainty.

- **Boundary overlap:** 3 windows straddle the 70/30 train/test boundary because `WINDOW_SIZE=24` and `STRIDE=6` produce overlapping windows. Given 208 test windows, this is a relatively small effect, but it makes the BATADAL temporal boundary somewhat softer than the SCADANet Week 10 split.

- **No variance estimate for the temporal point estimate:** the static BATADAL baseline uses 5-fold cross-validation with mean ± standard deviation, whereas each temporal result comes from a single chronological split. The 60/40, 70/30, and 80/20 sensitivity analysis provides additional evidence of stability, but it is not equivalent to cross-validation.

- **Threshold calibration limitation:** the corrected threshold selection now avoids test-set leakage by using OOF predictions from the training set only. However, the final calibrated threshold was selected from the same training data used to train the underlying model, through OOF predictions, rather than from an entirely independent validation dataset.

- **Calibration result is modest:** the corrected Max-F1 threshold improves F1 by **3.0 points**, rather than the larger improvement observed with the previous test-set-based calibration. The previous procedure used test labels to select the threshold and therefore benefited from information that should not be available during evaluation.

---

### 8. Takeaway for RQ2

The temporal evaluation reveals different generalization behaviors for the two datasets.

For **SCADANet**, temporal evaluation causes a substantial performance degradation, particularly in recall. The Week 10 analysis showed that `vuln_scan` is responsible for a large proportion of the false negatives, suggesting that the model has difficulty generalizing to later attack patterns even when those attack types were present during training.

For **BATADAL**, the temporal evaluation maintains high recall, but precision decreases because the default threshold produces additional false positives. The label-permutation test and split-boundary sensitivity analysis provide evidence that the result is not simply caused by memorization or one particular temporal split.

The threshold analysis further shows that **recall=1.0 does not mean perfect classification**. At the default threshold, all 21 attack windows are detected, but 12 normal windows are incorrectly classified as attacks.

After correcting the threshold-selection procedure to use training-only leave-one-fold-out predictions, the Max-F1 threshold improves F1 from **0.778 to 0.808** and reduces false positives from **12 to 10**, while still maintaining recall of **1.0**.

Overall, the results suggest two different issues:

**SCADANet shows limited temporal generalization to later attack patterns, while BATADAL shows evidence of temporal generalization but has a default operating point that produces unnecessary false positives.**

These should therefore be treated as different evaluation and modeling issues rather than as the same form of temporal degradation.

## Part X — Multi-Seed Stability Sweep (SCADANet Temporal Split, 5 seeds)

**Design:**
- The temporal train/test boundary (`SPLIT_WINDOW`), topology snapshots
  (`train_topology_edge_index`, `full_topology_edge_index`), and all
  hyperparameters are identical to Week 10 (Part M.2) -- nothing about the
  split itself changes here.
- 5 seeds, matching the seed count used for the BATADAL permutation test
  (Part R), so the two robustness checks in this notebook are on the same
  footing.
- `train_eval_scadanet_temporal` (Week 10) never called `torch.manual_seed`,
  so the reported 0.4923 recall was effectively an unseeded run. The
  function below (`train_eval_scadanet_temporal_seeded`) is otherwise
  identical but takes an explicit `seed` argument -- the Week 10 cell itself
  is left unmodified.
- **Scope note:** this sweep uses the default 0.5 decision threshold only,
  same as the original Week 10 result. It does not layer in Week 11's
  leave-one-fold-out threshold calibration -- that's a separate question
  (does calibration help this split?) from the one asked here (is the
  default-threshold result stable across seeds?).

````python
# X.1 -- Seeded version of the Week 10 temporal train/eval function
# Identical to train_eval_scadanet_temporal (Part M.2) except it takes an
# explicit seed and sets torch.manual_seed before model construction, so
# every part of the run that depends on the RNG (weight init AND any
# stochastic ops during training) is controlled.

def train_eval_scadanet_temporal_seeded(train_edge_index, eval_edge_index,
                                         query_edges, edge_attr, y,
                                         train_pos, test_pos, n_nodes,
                                         seed, epochs=400, lr=0.003, grad_clip=1.0):
    torch.manual_seed(seed)

    x = torch.ones((n_nodes, 4), dtype=torch.float32)
    train_data = Data(x=x, edge_index=train_edge_index)
    eval_data = Data(x=x, edge_index=eval_edge_index)

    train_idx = torch.tensor(train_pos, dtype=torch.long)
    test_idx = torch.tensor(test_pos, dtype=torch.long)
    train_y, test_y = y[train_idx], y[test_idx]

    class_counts = torch.bincount(train_y, minlength=2).float()
    class_weights = train_y.shape[0] / (2.0 * class_counts.clamp(min=1))

    model = EdgeClassifierWithAttr(in_dim=x.shape[1], edge_attr_dim=edge_attr.shape[1])
    optimizer = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=5e-4)

    model.train()
    for epoch in range(epochs):
        optimizer.zero_grad()
        out = model(train_data.x, train_data.edge_index,
                    query_edges[:, train_idx], edge_attr[train_idx])
        loss = F.cross_entropy(out, train_y, weight=class_weights)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=grad_clip)
        optimizer.step()

    model.eval()
    with torch.no_grad():
        test_pred = model(eval_data.x, eval_data.edge_index,
                           query_edges[:, test_idx], edge_attr[test_idx]).argmax(dim=1)
        acc = (test_pred == test_y).float().mean().item()
        tp = ((test_pred == 1) & (test_y == 1)).sum().item()
        fp = ((test_pred == 1) & (test_y == 0)).sum().item()
        fn = ((test_pred == 0) & (test_y == 1)).sum().item()
        tn = ((test_pred == 0) & (test_y == 0)).sum().item()
        precision = tp / (tp + fp) if (tp + fp) else 0.0
        recall = tp / (tp + fn) if (tp + fn) else 0.0
        f1 = 2 * precision * recall / (precision + recall) if (precision + recall) else 0.0

    return model, {
        "seed": seed, "test_accuracy": acc, "precision": precision,
        "recall": recall, "f1": f1,
        "confusion_matrix": {"tp": tp, "fp": fp, "fn": fn, "tn": tn},
    }


SEEDS = [42, 0, 1, 7, 123]  # 5 seeds, matching Part R (BATADAL permutation test)

seed_results = []
for s in SEEDS:
    _, m = train_eval_scadanet_temporal_seeded(
        train_topology_edge_index, full_topology_edge_index,
        scada_query_edges, scada_edge_attr, scada_y,
        train_pos, test_pos, n_nodes=len(all_ips), seed=s,
    )
    seed_results.append(m)
    acc_, prec_, rec_, f1_, cm_ = (m["test_accuracy"], m["precision"], m["recall"], m["f1"], m["confusion_matrix"])
    print(f"seed={s:<5} acc={acc_:.4f}  prec={prec_:.4f}  recall={rec_:.4f}  f1={f1_:.4f}  cm={cm_}")
````

## Part Y — Feature Drift on `vuln_scan`, Train Period vs Test Period

Restricted to `vuln_scan` flows only, comparing train-period vs test-period
values of the three features Week 9's Integrated Gradients ranked as the
model's strongest decision drivers (Part K findings: `Protocol_TCP` 0.273,
`Tcp_flags_reset_Set` 0.154, `frame_len` 0.098). Scoping to `vuln_scan` only
(rather than all traffic) keeps the comparison focused on the flows the
model is actually failing on, rather than mixing in drift from traffic types
where recall isn't a problem.

`Protocol_TCP` and `Tcp_flags_reset_Set` are one-hot indicator columns
(0/1), so they're compared as **rates** (train rate vs test rate, with a
two-proportion z-test) rather than with a KS-test, which assumes a
continuous distribution and isn't the right tool for a binary column.
`frame_len` is numeric and uses the raw (unnormalized) values from
`scada_df` -- not the z-scored version in `scada_edge_attr`, since z-scoring
was fit on the full population and would obscure the raw shift -- compared
with a KS-test and PSI.

````python
# Y.1 -- Isolate vuln_scan rows within the train period and the test period
vuln_train_mask = train_types_arr == "vuln_scan"
vuln_test_mask = test_types_arr == "vuln_scan"

vuln_train_pos = train_pos[vuln_train_mask]
vuln_test_pos = test_pos[vuln_test_mask]

print(f"vuln_scan rows available for drift analysis: "
      f"{len(vuln_train_pos)} (train period), {len(vuln_test_pos)} (test period)")

if len(vuln_train_pos) < 10 or len(vuln_test_pos) < 10:
    print("\nWARNING: fewer than 10 vuln_scan rows on one side -- KS-test and "
          "z-test results below will be low-power; treat p-values as indicative "
          "only, not confirmatory.")
````

````python
# Y.2 -- Binary IG features: rate comparison + two-proportion z-test
from scipy.stats import norm

protocol_tcp_idx = edge_feature_names.index("Protocol_TCP")
tcp_reset_idx = edge_feature_names.index("Tcp_flags_reset_Set")

def two_proportion_ztest(x1, n1, x2, n2):
    if n1 == 0 or n2 == 0:
        return float("nan"), float("nan")
    p1, p2 = x1 / n1, x2 / n2
    p_pool = (x1 + x2) / (n1 + n2)
    se = np.sqrt(p_pool * (1 - p_pool) * (1 / n1 + 1 / n2))
    if se == 0:
        return 0.0, 1.0
    z = (p1 - p2) / se
    pval = 2 * (1 - norm.cdf(abs(z)))
    return z, pval

binary_drift_results = {}

print("=" * 78)
print("BINARY IG-FEATURE RATE COMPARISON -- vuln_scan ONLY, train vs test period")
print("=" * 78)
for name, idx in [("Protocol_TCP", protocol_tcp_idx), ("Tcp_flags_reset_Set", tcp_reset_idx)]:
    train_vals = scada_edge_attr[vuln_train_pos, idx].numpy()
    test_vals = scada_edge_attr[vuln_test_pos, idx].numpy()
    x1, n1 = float(train_vals.sum()), len(train_vals)
    x2, n2 = float(test_vals.sum()), len(test_vals)
    z, pval = two_proportion_ztest(x1, n1, x2, n2)
    rate1 = x1 / n1 if n1 else float("nan")
    rate2 = x2 / n2 if n2 else float("nan")
    print(f"{name:<22} train_rate={rate1:.3f} (n={n1})   "
          f"test_rate={rate2:.3f} (n={n2})   z={z:.3f}   p={pval:.4f}")
    binary_drift_results[name] = {
        "train_rate": rate1, "test_rate": rate2,
        "n_train": n1, "n_test": n2, "z": float(z), "p_value": float(pval),
    }
````

````python
# Y.3 -- frame_len (raw, unnormalized): KS-test + PSI + histogram overlay
from scipy.stats import ks_2samp

frame_len_train = scada_df["frame_len"].iloc[vuln_train_pos].astype(float).to_numpy()
frame_len_test = scada_df["frame_len"].iloc[vuln_test_pos].astype(float).to_numpy()

ks_stat, ks_pval = ks_2samp(frame_len_train, frame_len_test)

def psi(expected, actual, buckets=10):
    # Fix (review): the previous version built percentile edges from
    # `expected` (train) alone and bailed out to NaN whenever those edges
    # collapsed to fewer than 3 unique values after np.unique(). frame_len
    # has heavy ties (median 60 in both periods -- few distinct raw byte
    # values for scan probes), so most percentile breakpoints landed on the
    # same value and the function returned NaN instead of degrading
    # gracefully. Fix: build edges from the POOLED train+test data (more
    # representative of where real breaks exist), and skip bins with zero
    # count on both sides rather than aborting the whole computation.
    expected = expected[~np.isnan(expected)]
    actual = actual[~np.isnan(actual)]
    pooled = np.concatenate([expected, actual])

    breakpoints = np.linspace(0, 100, buckets + 1)
    edges = np.unique(np.percentile(pooled, breakpoints))
    if len(edges) < 2:
        # Only possible if the pooled data is a single repeated value --
        # a genuine degenerate case, not the tie-collapse bug above.
        return float("nan")
    edges = edges.copy()
    edges[0] -= 1e-6
    edges[-1] += 1e-6

    exp_counts, _ = np.histogram(expected, bins=edges)
    act_counts, _ = np.histogram(actual, bins=edges)

    # Skip bins with zero observations in BOTH distributions -- these are
    # artifacts of low-cardinality data producing extra bin edges, not real
    # distributional information, and would otherwise contribute spurious
    # near-zero terms via the clip() floor below.
    keep = (exp_counts + act_counts) > 0
    if keep.sum() == 0:
        return float("nan")

    exp_pct = np.clip(exp_counts[keep] / len(expected), 1e-4, None)
    act_pct = np.clip(act_counts[keep] / len(actual), 1e-4, None)
    return float(np.sum((act_pct - exp_pct) * np.log(act_pct / exp_pct)))

frame_len_psi = psi(frame_len_train, frame_len_test)

print("=" * 70)
print("FRAME_LEN DRIFT -- vuln_scan ONLY, train vs test period (raw values)")
print("=" * 70)
print(f"train: n={len(frame_len_train)}  mean={frame_len_train.mean():.1f}  "
      f"median={np.median(frame_len_train):.1f}  std={frame_len_train.std():.1f}")
print(f"test:  n={len(frame_len_test)}  mean={frame_len_test.mean():.1f}  "
      f"median={np.median(frame_len_test):.1f}  std={frame_len_test.std():.1f}")
print(f"\nKS statistic={ks_stat:.4f}  p-value={ks_pval:.4g}")
print(f"PSI={frame_len_psi:.4f}  (guide: <0.1 no meaningful shift, "
      f"0.1-0.25 moderate shift, >0.25 major shift)")

fig, ax = plt.subplots(figsize=(7, 5))
ax.hist(frame_len_train, bins=30, alpha=0.5, density=True,
        label=f"train period (n={len(frame_len_train)})", color="#4C72B0")
ax.hist(frame_len_test, bins=30, alpha=0.5, density=True,
        label=f"test period (n={len(frame_len_test)})", color="#DD8452")
ax.set_xlabel("frame_len (raw bytes)")
ax.set_ylabel("density")
ax.set_title("vuln_scan frame_len distribution: train vs test period")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("week12_vuln_scan_frame_len_drift.png", dpi=150)
plt.show()

frame_len_drift_results = {
    "ks_statistic": float(ks_stat), "ks_p_value": float(ks_pval), "psi": frame_len_psi,
    "train_mean": float(frame_len_train.mean()), "test_mean": float(frame_len_test.mean()),
    "train_median": float(np.median(frame_len_train)), "test_median": float(np.median(frame_len_test)),
    "n_train": len(frame_len_train), "n_test": len(frame_len_test),
}
````

## **Part Z — Findings (Parts W–Y)**

### 1. `vuln_scan` base rate (Part W)

The `vuln_scan` rate is **0.83% in the training period** (3,093 / 374,389 flows), compared with **23.54% in the test period** (37,777 / 160,452 flows). Thus, the test-period `vuln_scan` rate is **28.50× higher** than the training-period rate.

This represents a substantial class-distribution shift between the two temporal periods. The model is therefore trained with very few `vuln_scan` examples relative to the later test period, while evaluation is performed on a period in which `vuln_scan` is substantially more prevalent. This large change in class prevalence is an important characteristic of the temporal setting and demonstrates that the test distribution differs considerably from the training distribution.

### 2. Seed stability (Part X)

**Note (post-review correction):** the table below was originally computed with a bug in the `seed=7` run (results below reflect the corrected re-run; see notebook history / `week12_metricsnew.json` for the before/after). The corrected numbers are used throughout this section.

Across five random seeds, the temporal-split results were:

| Metric    |   Mean |    Std |    Min |    Max |
| --------- | -----: | -----: | -----: | -----: |
| Accuracy  | 0.7092 | 0.0100 | 0.6968 | 0.7175 |
| Precision | 0.8708 | 0.0061 | 0.8627 | 0.8767 |
| Recall    | 0.5168 | 0.0179 | 0.4948 | 0.5326 |
| F1        | 0.6485 | 0.0158 | 0.6291 | 0.6620 |

The Week 10 unseeded result was **recall = 0.4923** and **F1 = 0.6276**. These values remain very close to the seeded results: the Week 10 recall is only **0.0025 below the seeded minimum**, while the F1 is **0.0015 below the seeded minimum** (the min/max themselves are unaffected by the seed=7 correction, since seed 7 was never the extreme run). The five seeded recall values remain within the narrow range of **0.4948–0.5326**.

The relatively small variation across seeds demonstrates that the temporal performance degradation is **stable across different random initializations** and is not the result of an unusually unfavorable training seed. The Week 10 result is therefore representative of the observed temporal-split behavior.

**Clustering note:** the five seeds are not smoothly spread around the mean -- they split into two groups. Seeds 42, 0, and 7 all land around **0.717 accuracy / 0.53 recall**, while seeds 1 and 123 land around **0.697 accuracy / 0.495 recall**. This bimodal pattern (rather than unimodal noise around 0.709) is worth a follow-up look -- e.g. whether the low-accuracy seeds happen to produce an unluckier train/test split of `vuln_scan` flows specifically, given how small and imbalanced that class is (3,093 train examples). The overall stability conclusion above still holds either way (both clusters sit well below the static IP-held-out numbers and both clearly demonstrate the same degradation), but the two-cluster structure is a more precise description of the spread than "std ≈ 0.01" alone conveys.

### 3. Feature drift on `vuln_scan` (Part Y)

The three features identified as important by Week 9 Integrated Gradients were examined separately for `vuln_scan` flows in the training and test periods.

**`Protocol_TCP`:**
The training rate was **0.982**, compared with **0.988** in the test period. Although the difference is statistically significant (**z = -3.026, p = 0.0025**), the absolute difference is only **0.006 (0.6 percentage points)**. This indicates a statistically detectable but practically very small change.

**`Tcp_flags_reset_Set`:**
The training and test rates were both approximately **0.001**. The difference was not statistically significant (**z = -0.229, p = 0.8188**), indicating no evidence of meaningful distributional change.

**`frame_len`:**
The training period had a mean of **63.4** and median of **60.0**, while the test period had a mean of **62.2** and median of **60.0**. The KS statistic was only **0.0062** with **p = 0.9999**, providing no evidence of a meaningful distributional shift.

The corrected PSI (fixed binning: edges built from the pooled train+test distribution, zero-count bins skipped rather than aborting the computation -- see Y.3) is **PSI = 0.0000**, consistent with the KS result: no meaningful distributional shift in `frame_len` between the training and test periods. (The original, buggy version of this function returned `NaN` here -- root cause: `frame_len` has heavy ties for `vuln_scan` traffic, median 60 in both periods, so percentile bin edges built from train data alone collapsed below the minimum required for the histogram comparison and the function aborted instead of degrading gracefully.)

Overall, the three strongest IG features from Week 9 **remain highly consistent between the training and test `vuln_scan` traffic**. `Protocol_TCP` shows only a 0.6-percentage-point difference, while `Tcp_flags_reset_Set` and `frame_len` show no meaningful evidence of drift.

Thus, the temporal recall degradation is **not attributable to substantial drift in the model's three strongest identified content features**.

### 4. Takeaway for RQ2

The Week 10 temporal recall degradation is **real and stable across random seeds**. Across five seeded runs, recall remained within **0.4948–0.5326**, with a mean of **0.5168**, while the Week 10 unseeded result was **0.4923**. The close agreement between the seeded and unseeded results demonstrates that the observed temporal degradation is a consistent property of the temporal evaluation rather than random training variation.

The temporal evaluation also reveals a substantial change in `vuln_scan` prevalence, increasing from **0.83% in the training period to 23.54% in the test period**, representing a **28.50× increase**. This establishes a major difference in class prevalence between the training and test periods.

At the same time, the three strongest features identified through Week 9 Integrated Gradients—`Protocol_TCP`, `Tcp_flags_reset_Set`, and `frame_len`—remain highly consistent across the two periods, with no practically meaningful drift observed.

Therefore, the RQ2 finding is:

> **The temporal evaluation demonstrates that model performance degrades substantially when evaluated on later SCADANet traffic, and this degradation is stable across random seeds. The temporal setting is characterized by a major 28.50× increase in `vuln_scan` prevalence, confirming a substantial change between the training and test distributions. However, the model's three strongest previously identified content features remain largely stable across the two periods, indicating that the observed temporal degradation is not driven by meaningful drift in these features.**

**RQ2 conclusion: The temporal evaluation successfully demonstrates stable degradation under a substantially changed future traffic distribution, while the explainability and feature-drift analysis shows that this degradation cannot be attributed to changes in the model's three strongest identified features.**
